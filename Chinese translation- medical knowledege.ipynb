# PREPROCESSING

```python

import os
import re
import cv2
import numpy as np
import pandas as pd
import unicodedata

from pathlib import Path
from sklearn.preprocessing import StandardScaler
from IPython.display import display

# ================================================================
# STEP 2: MULTIMODAL TCM DATASET PREPROCESSING
# ================================================================

print("=" * 110)
print("STEP 2: MULTIMODAL TRADITIONAL CHINESE MEDICINE DATASET PREPROCESSING")
print("=" * 110)

# ------------------------------------------------
# 1. INPUT PATHS
# ------------------------------------------------

BASE_DIR = r"D:\Implementation\2_Medicine Knowledge\Dataset"

IMAGE_DIR = os.path.join(BASE_DIR, "Images")

AUDIO_CSV = os.path.join(
    BASE_DIR, "Herb_Audio_Features.csv"
)

TEXT_CSV = os.path.join(
    BASE_DIR, "Herb_Text_Translation.csv"
)

# ------------------------------------------------
# 2. OUTPUT PATHS
# ------------------------------------------------

OUTPUT_DIR = os.path.join(BASE_DIR, "Preprocessed")

PREPROCESSED_IMAGE_DIR = os.path.join(
    OUTPUT_DIR, "Preprocessed_Images"
)

PREPROCESSED_AUDIO_CSV = os.path.join(
    OUTPUT_DIR, "Audio_Preprocessed.csv"
)

PREPROCESSED_TEXT_CSV = os.path.join(
    OUTPUT_DIR, "Text_Preprocessed.csv"
)

os.makedirs(PREPROCESSED_IMAGE_DIR, exist_ok=True)

# ------------------------------------------------
# 3. PARAMETERS
# ------------------------------------------------

IMAGE_SIZE = (224, 224)

IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png",
    ".bmp", ".tif", ".tiff", ".webp"
}

# ------------------------------------------------
# 4. LOAD CSV DATASETS
# ------------------------------------------------

print("\n[1] LOADING ORIGINAL DATASETS")
print("-" * 110)

def load_csv_safely(path):
    for encoding in ["utf-8", "utf-8-sig", "gbk", "latin1"]:
        try:
            return pd.read_csv(path, encoding=encoding)
        except UnicodeDecodeError:
            continue
    raise ValueError(f"Could not read CSV: {path}")

audio_df = load_csv_safely(AUDIO_CSV)
text_df = load_csv_safely(TEXT_CSV)

print("Original Audio Dataset Shape:", audio_df.shape)
print("Original Text Dataset Shape :", text_df.shape)

# ------------------------------------------------
# 5. IMAGE PREPROCESSING
# ------------------------------------------------

print("\n[2] IMAGE PREPROCESSING")
print("-" * 110)

def preprocess_image(image_path, output_path):

    # Read image
    image = cv2.imread(image_path)

    if image is None:
        return False

    # Convert BGR to RGB
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Resize image
    image = cv2.resize(
        image,
        IMAGE_SIZE,
        interpolation=cv2.INTER_AREA
    )

    # Noise filtering
    image = cv2.bilateralFilter(
        image,
        d=5,
        sigmaColor=50,
        sigmaSpace=50
    )

    # Convert RGB to LAB color space
    lab_image = cv2.cvtColor(
        image,
        cv2.COLOR_RGB2LAB
    )

    # Apply CLAHE to luminance channel
    l_channel, a_channel, b_channel = cv2.split(lab_image)

    clahe = cv2.createCLAHE(
        clipLimit=2.0,
        tileGridSize=(8, 8)
    )

    enhanced_l = clahe.apply(l_channel)

    # Merge enhanced channels
    enhanced_lab = cv2.merge(
        [enhanced_l, a_channel, b_channel]
    )

    # Convert back to RGB
    processed_image = cv2.cvtColor(
        enhanced_lab,
        cv2.COLOR_LAB2RGB
    )

    # Save preprocessed image
    processed_bgr = cv2.cvtColor(
        processed_image,
        cv2.COLOR_RGB2BGR
    )

    os.makedirs(
        os.path.dirname(output_path),
        exist_ok=True
    )

    success = cv2.imwrite(
        output_path,
        processed_bgr,
        [cv2.IMWRITE_JPEG_QUALITY, 100]
    )

    return success


image_count = 0
failed_images = []

for root, dirs, files in os.walk(IMAGE_DIR):

    for filename in files:

        extension = Path(filename).suffix.lower()

        if extension not in IMAGE_EXTENSIONS:
            continue

        input_path = os.path.join(root, filename)

        # Preserve original herb-folder structure
        relative_path = os.path.relpath(
            input_path,
            IMAGE_DIR
        )

        # Save every processed image as JPG
        relative_path = str(
            Path(relative_path).with_suffix(".jpg")
        )

        output_path = os.path.join(
            PREPROCESSED_IMAGE_DIR,
            relative_path
        )

        try:
            success = preprocess_image(
                input_path,
                output_path
            )

            if success:
                image_count += 1
            else:
                failed_images.append(input_path)

        except Exception as error:
            failed_images.append(input_path)
            print(f"Image processing error: {filename} - {error}")

print("Original Image Directory :", IMAGE_DIR)
print("Preprocessed Image Folder:", PREPROCESSED_IMAGE_DIR)
print("Successfully Processed   :", image_count)
print("Failed Images            :", len(failed_images))

# ------------------------------------------------
# 6. AUDIO FEATURE PREPROCESSING
# ------------------------------------------------

print("\n[3] AUDIO FEATURE PREPROCESSING")
print("-" * 110)

audio_processed = audio_df.copy()

# Clean column names
audio_processed.columns = (
    audio_processed.columns.astype(str).str.strip()
)

# Clean audio filename and herb name
for column in ["audio_name", "herb_name"]:
    if column in audio_processed.columns:
        audio_processed[column] = (
            audio_processed[column]
            .astype("string")
            .str.strip()
        )

# Convert numerical columns
audio_numeric_columns = [
    "ID",
    "duration_sec",
    "sample_rate",
    "rms_energy",
    "zero_crossing_rate",
    "spectral_centroid",
    "spectral_bandwidth"
]

for column in audio_numeric_columns:

    if column in audio_processed.columns:
        audio_processed[column] = pd.to_numeric(
            audio_processed[column],
            errors="coerce"
        )

# Remove invalid records
audio_processed = audio_processed.dropna(
    subset=[
        column for column in audio_numeric_columns
        if column in audio_processed.columns
    ] + [
        column for column in ["audio_name", "herb_name"]
        if column in audio_processed.columns
    ]
).copy()

# Validate positive duration and sample rate
if "duration_sec" in audio_processed.columns:
    audio_processed = audio_processed[
        audio_processed["duration_sec"] > 0
    ]

if "sample_rate" in audio_processed.columns:
    audio_processed = audio_processed[
        audio_processed["sample_rate"] > 0
    ]

# Standardize continuous audio features
audio_feature_columns = [
    "duration_sec",
    "rms_energy",
    "zero_crossing_rate",
    "spectral_centroid",
    "spectral_bandwidth"
]

audio_feature_columns = [
    column for column in audio_feature_columns
    if column in audio_processed.columns
]

scaler = StandardScaler()

scaled_values = scaler.fit_transform(
    audio_processed[audio_feature_columns]
)

# Preserve original features and add standardized features
for index, column in enumerate(audio_feature_columns):

    audio_processed[column + "_scaled"] = (
        scaled_values[:, index]
    )

# Save preprocessed audio CSV
audio_processed.to_csv(
    PREPROCESSED_AUDIO_CSV,
    index=False,
    encoding="utf-8-sig"
)

print("Audio Records Before Preprocessing:", len(audio_df))
print("Audio Records After Preprocessing :", len(audio_processed))
print("Audio Original Features           :", len(audio_feature_columns))
print("Standardized Features Added       :", len(audio_feature_columns))
print("Audio Output Shape                :", audio_processed.shape)
print("Saved Audio CSV                   :", PREPROCESSED_AUDIO_CSV)

# ------------------------------------------------
# 7. TEXT TRANSLATION PREPROCESSING
# ------------------------------------------------

print("\n[4] TEXT TRANSLATION PREPROCESSING")
print("-" * 110)

text_processed = text_df.copy()

# Clean column names
text_processed.columns = (
    text_processed.columns.astype(str).str.strip()
)

# Clean text while preserving Chinese and English content
def clean_text(value):

    if pd.isna(value):
        return ""

    value = unicodedata.normalize(
        "NFC",
        str(value)
    )

    # Replace line breaks and tabs with spaces
    value = re.sub(
        r"[\r\n\t]+",
        " ",
        value
    )

    # Remove repeated whitespace
    value = re.sub(
        r"\s+",
        " ",
        value
    )

    return value.strip()


text_columns = [
    "image_name",
    "herb_usage_china",
    "english_translation"
]

for column in text_columns:

    if column in text_processed.columns:

        text_processed[column] = (
            text_processed[column]
            .apply(clean_text)
        )

# Convert ID and translation score to numeric
for column in ["ID", "translation_score"]:

    if column in text_processed.columns:

        text_processed[column] = pd.to_numeric(
            text_processed[column],
            errors="coerce"
        )

# Remove records with missing essential information
required_columns = [
    column for column in [
        "ID",
        "image_name",
        "herb_usage_china",
        "english_translation",
        "translation_score"
    ]
    if column in text_processed.columns
]

text_processed = text_processed.dropna(
    subset=required_columns
).copy()

# Remove empty text records
text_processed = text_processed[
    (text_processed["herb_usage_china"].str.len() > 0) &
    (text_processed["english_translation"].str.len() > 0)
].copy()

# Validate translation score range
if "translation_score" in text_processed.columns:

    text_processed = text_processed[
        text_processed["translation_score"].between(0, 1)
    ].copy()

# Save preprocessed text CSV
text_processed.to_csv(
    PREPROCESSED_TEXT_CSV,
    index=False,
    encoding="utf-8-sig"
)

print("Text Records Before Preprocessing:", len(text_df))
print("Text Records After Preprocessing :", len(text_processed))
print("Text Columns                     :", len(text_processed.columns))
print("Text Output Shape                :", text_processed.shape)
print("Saved Text CSV                   :", PREPROCESSED_TEXT_CSV)

# ------------------------------------------------
# 8. IMAGE AND TEXT RECORD MATCHING
# ------------------------------------------------

print("\n[5] IMAGE AND TEXT RECORD MATCHING")
print("-" * 110)

processed_image_names = {
    os.path.basename(os.path.relpath(path, PREPROCESSED_IMAGE_DIR)).lower()
    for root, dirs, files in os.walk(PREPROCESSED_IMAGE_DIR)
    for path in [
        os.path.join(root, filename)
        for filename in files
        if Path(filename).suffix.lower() == ".jpg"
    ]
}

text_image_names = set(
    text_processed["image_name"]
    .astype(str)
    .str.strip()
    .str.lower()
)

matched_images = processed_image_names.intersection(
    text_image_names
)

missing_images = text_image_names - processed_image_names

print("Preprocessed Images Found :", len(processed_image_names))
print("Text Image References     :", len(text_image_names))
print("Matching Image Names      :", len(matched_images))
print("Missing Image References  :", len(missing_images))

if missing_images:
    print("\nFirst 10 Missing Image References:")
    for name in list(sorted(missing_images))[:10]:
        print(name)

# ------------------------------------------------
# 9. HERB-WISE IMAGE COUNT
# ------------------------------------------------

print("\n[6] HERB-WISE PREPROCESSED IMAGE COUNT")
print("-" * 110)

herb_counts = {}

for herb_folder in sorted(os.listdir(PREPROCESSED_IMAGE_DIR)):

    herb_path = os.path.join(
        PREPROCESSED_IMAGE_DIR,
        herb_folder
    )

    if os.path.isdir(herb_path):

        count = sum(
            1 for filename in os.listdir(herb_path)
            if Path(filename).suffix.lower() == ".jpg"
        )

        herb_counts[herb_folder] = count

for herb, count in herb_counts.items():
    print(f"{herb:<50} : {count}")

print("\nTotal Herb Categories:", len(herb_counts))

# ------------------------------------------------
# 10. FINAL PREPROCESSING SUMMARY
# ------------------------------------------------

print("\n" + "=" * 110)
print("STEP 2: MULTIMODAL DATASET PREPROCESSING COMPLETED")
print("=" * 110)

print("\nIMAGE PREPROCESSING")
print("-" * 110)
print("Original Images       :", image_count + len(failed_images))
print("Successfully Processed:", image_count)
print("Failed Images         :", len(failed_images))
print("Image Size            :", f"{IMAGE_SIZE[0]} x {IMAGE_SIZE[1]}")
print("Output Directory      :", PREPROCESSED_IMAGE_DIR)

print("\nAUDIO PREPROCESSING")
print("-" * 110)
print("Original Audio Records:", len(audio_df))
print("Processed Audio Rows  :", len(audio_processed))
print("Processed Audio Shape :", audio_processed.shape)
print("Output CSV            :", PREPROCESSED_AUDIO_CSV)

print("\nTEXT PREPROCESSING")
print("-" * 110)
print("Original Text Records :", len(text_df))
print("Processed Text Rows   :", len(text_processed))
print("Processed Text Shape  :", text_processed.shape)
print("Output CSV            :", PREPROCESSED_TEXT_CSV)
print("\nAll preprocessing operations completed.")
```

# Feature Extraction

```python

import os
import re
import cv2
import numpy as np
import pandas as pd
import tensorflow as tf
import unicodedata

from pathlib import Path
from gensim.models import Word2Vec
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.applications.resnet50 import preprocess_input
from tensorflow.keras.models import Model

# ================================================================
# STEP 3: MULTIMODAL FEATURE EXTRACTION
# ================================================================

print("=" * 110)
print("STEP 3: MULTIMODAL TRADITIONAL CHINESE MEDICINE FEATURE EXTRACTION")
print("=" * 110)

# ------------------------------------------------
# 1. DEFINE INPUT AND OUTPUT PATHS
# ------------------------------------------------

BASE_DIR = r"D:\Implementation\2_Medicine Knowledge\Dataset"

PREPROCESSED_DIR = os.path.join(
    BASE_DIR, "Preprocessed"
)

IMAGE_DIR = os.path.join(
    PREPROCESSED_DIR, "Preprocessed_Images"
)

AUDIO_CSV = os.path.join(
    PREPROCESSED_DIR, "Audio_Preprocessed.csv"
)

TEXT_CSV = os.path.join(
    PREPROCESSED_DIR, "Text_Preprocessed.csv"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR, "Feature_Extraction"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

IMAGE_FEATURE_CSV = os.path.join(
    OUTPUT_DIR, "Image_Extracted_Features.csv"
)

AUDIO_FEATURE_CSV = os.path.join(
    OUTPUT_DIR, "Audio_Extracted_Features.csv"
)

TEXT_FEATURE_CSV = os.path.join(
    OUTPUT_DIR, "Text_Extracted_Features.csv"
)

# ------------------------------------------------
# 2. CONFIGURATION
# ------------------------------------------------

IMAGE_SIZE = (224, 224)
IMAGE_BATCH_SIZE = 32

WORD2VEC_DIM = 300
WORD2VEC_WINDOW = 5
WORD2VEC_MIN_COUNT = 1
WORD2VEC_EPOCHS = 30

RANDOM_SEED = 42

np.random.seed(RANDOM_SEED)
tf.random.set_seed(RANDOM_SEED)

# ------------------------------------------------
# 3. LOAD PREPROCESSED DATASETS
# ------------------------------------------------

print("\n[1] LOADING PREPROCESSED DATASETS")
print("-" * 110)

audio_df = pd.read_csv(AUDIO_CSV, encoding="utf-8-sig")
text_df = pd.read_csv(TEXT_CSV, encoding="utf-8-sig")

print("Audio Dataset Shape:", audio_df.shape)
print("Text Dataset Shape :", text_df.shape)

# ------------------------------------------------
# 4. IMAGE FEATURE EXTRACTION USING RESNET50
# ------------------------------------------------

print("\n[2] IMAGE FEATURE EXTRACTION - RESNET50")
print("-" * 110)

# Load pretrained ResNet50
# include_top=False removes the classification layer.
# pooling='avg' produces 2048-dimensional feature vectors.

resnet_model = ResNet50(
    weights="imagenet",
    include_top=False,
    pooling="avg",
    input_shape=(224, 224, 3)
)

print("ResNet50 Model Loaded Successfully")
print("Feature Dimension:", 2048)

# Create filename-to-ID mapping from text dataset
image_id_mapping = {}

for _, row in text_df.iterrows():

    image_name = str(row["image_name"]).strip().lower()

    image_id_mapping[image_name] = row["ID"]

# Collect preprocessed image paths
image_paths = []

for root, dirs, files in os.walk(IMAGE_DIR):

    for filename in files:

        if Path(filename).suffix.lower() in [
            ".jpg", ".jpeg", ".png", ".bmp"
        ]:

            image_paths.append(
                os.path.join(root, filename)
            )

image_paths = sorted(image_paths)

print("Total Preprocessed Images:", len(image_paths))

image_feature_records = []

# Process images in batches
for start in range(0, len(image_paths), IMAGE_BATCH_SIZE):

    batch_paths = image_paths[
        start:start + IMAGE_BATCH_SIZE
    ]

    batch_images = []
    valid_paths = []

    for image_path in batch_paths:

        try:

            image = cv2.imread(image_path)

            if image is None:
                continue

            image = cv2.cvtColor(
                image,
                cv2.COLOR_BGR2RGB
            )

            image = cv2.resize(
                image,
                IMAGE_SIZE
            )

            image = image.astype(np.float32)

            image = preprocess_input(image)

            batch_images.append(image)
            valid_paths.append(image_path)

        except Exception as error:

            print(
                "Image Loading Error:",
                os.path.basename(image_path),
                error
            )

    if len(batch_images) == 0:
        continue

    batch_images = np.array(
        batch_images,
        dtype=np.float32
    )

    # Extract ResNet50 features
    batch_features = resnet_model.predict(
        batch_images,
        verbose=0
    )

    # Store each image feature vector
    for image_path, features in zip(
        valid_paths,
        batch_features
    ):

        image_name = os.path.basename(image_path)

        image_id = image_id_mapping.get(
            image_name.lower(),
            np.nan
        )

        record = {
            "image_name": image_name,
            "ID": image_id
        }

        for index, value in enumerate(features):

            record[
                f"Image_Feature_{index + 1:04d}"
            ] = float(value)

        image_feature_records.append(record)

    print(
        f"Processed {min(start + len(batch_paths), len(image_paths))}"
        f" / {len(image_paths)} images"
    )

# Create image feature dataframe
image_features_df = pd.DataFrame(
    image_feature_records
)

# Save image feature CSV
image_features_df.to_csv(
    IMAGE_FEATURE_CSV,
    index=False,
    encoding="utf-8-sig"
)

print("\nIMAGE FEATURE EXTRACTION SUMMARY")
print("-" * 110)

print("Processed Images :", len(image_features_df))
print("Feature Dimension:", 2048)
print("Output Shape     :", image_features_df.shape)
print("Saved CSV        :", IMAGE_FEATURE_CSV)

# ------------------------------------------------
# 5. AUDIO FEATURE EXTRACTION
# ------------------------------------------------

print("\n[3] AUDIO FEATURE EXTRACTION")
print("-" * 110)

audio_metadata_columns = [
    "audio_name",
    "ID",
    "herb_name"
]

audio_numeric_columns = [
    "duration_sec_scaled",
    "rms_energy_scaled",
    "zero_crossing_rate_scaled",
    "spectral_centroid_scaled",
    "spectral_bandwidth_scaled"
]

# Check standardized feature columns
missing_audio_columns = [
    col for col in audio_numeric_columns
    if col not in audio_df.columns
]

if missing_audio_columns:

    raise ValueError(
        "Missing standardized audio columns: "
        + str(missing_audio_columns)
    )

# Select metadata and extracted numerical features
audio_features_df = audio_df[
    audio_metadata_columns + audio_numeric_columns
].copy()

# Rename extracted features
audio_features_df.rename(
    columns={
        "duration_sec_scaled": "Audio_Feature_01",
        "rms_energy_scaled": "Audio_Feature_02",
        "zero_crossing_rate_scaled": "Audio_Feature_03",
        "spectral_centroid_scaled": "Audio_Feature_04",
        "spectral_bandwidth_scaled": "Audio_Feature_05"
    },
    inplace=True
)

# Save audio feature CSV
audio_features_df.to_csv(
    AUDIO_FEATURE_CSV,
    index=False,
    encoding="utf-8-sig"
)

print("Audio Records       :", len(audio_features_df))
print("Audio Feature Count :", 5)
print("Audio Feature Shape :", audio_features_df.shape)
print("Saved CSV           :", AUDIO_FEATURE_CSV)

# ------------------------------------------------
# 6. TEXT FEATURE EXTRACTION USING WORD2VEC
# ------------------------------------------------

print("\n[4] TEXT FEATURE EXTRACTION - WORD2VEC")
print("-" * 110)

try:
    import jieba

    print("Chinese Tokenizer: Jieba")

    def tokenize_chinese(text):

        text = str(text).strip()

        tokens = [
            word.strip()
            for word in jieba.lcut(text)
            if word.strip()
        ]

        return tokens

except ImportError:

    print(
        "Jieba not installed. "
        "Using Chinese character-level tokenization."
    )

    def tokenize_chinese(text):

        text = str(text).strip()

        return [
            char for char in text
            if not char.isspace()
        ]


def tokenize_english(text):

    text = unicodedata.normalize(
        "NFKC",
        str(text).lower()
    )

    # Retain alphabetic words and digits
    tokens = re.findall(
        r"[a-z]+(?:'[a-z]+)?|\d+",
        text
    )

    return tokens


# Prepare combined bilingual token sequences.
# Each record contributes its Chinese and English tokens.

sentences = []

for _, row in text_df.iterrows():

    chinese_tokens = tokenize_chinese(
        row["herb_usage_china"]
    )

    english_tokens = tokenize_english(
        row["english_translation"]
    )

    combined_tokens = (
        chinese_tokens + english_tokens
    )

    if len(combined_tokens) > 0:

        sentences.append(combined_tokens)

print("Total Text Records:", len(sentences))

# Train Word2Vec
word2vec_model = Word2Vec(
    sentences=sentences,
    vector_size=WORD2VEC_DIM,
    window=WORD2VEC_WINDOW,
    min_count=WORD2VEC_MIN_COUNT,
    workers=1,
    sg=1,
    seed=RANDOM_SEED,
    epochs=WORD2VEC_EPOCHS
)

print("Word2Vec Model Trained")
print("Vocabulary Size:", len(word2vec_model.wv))
print("Embedding Dimension:", WORD2VEC_DIM)


# Average token embeddings to produce a fixed-size vector
def generate_sentence_embedding(tokens, model):

    valid_vectors = [
        model.wv[token]
        for token in tokens
        if token in model.wv
    ]

    if len(valid_vectors) == 0:

        return np.zeros(
            WORD2VEC_DIM,
            dtype=np.float32
        )

    return np.mean(
        valid_vectors,
        axis=0
    ).astype(np.float32)


text_feature_records = []

for _, row in text_df.iterrows():

    chinese_tokens = tokenize_chinese(
        row["herb_usage_china"]
    )

    english_tokens = tokenize_english(
        row["english_translation"]
    )

    combined_tokens = (
        chinese_tokens + english_tokens
    )

    embedding = generate_sentence_embedding(
        combined_tokens,
        word2vec_model
    )

    record = {
        "image_name": row["image_name"],
        "ID": row["ID"],
        "translation_score": row["translation_score"]
    }

    for index, value in enumerate(embedding):

        record[
            f"Text_Feature_{index + 1:03d}"
        ] = float(value)

    text_feature_records.append(record)

# Create text feature dataframe
text_features_df = pd.DataFrame(
    text_feature_records
)

# Save text feature CSV
text_features_df.to_csv(
    TEXT_FEATURE_CSV,
    index=False,
    encoding="utf-8-sig"
)

print("\nTEXT FEATURE EXTRACTION SUMMARY")
print("-" * 110)

print("Text Records       :", len(text_features_df))
print("Text Feature Count :", WORD2VEC_DIM)
print("Text Feature Shape :", text_features_df.shape)
print("Saved CSV          :", TEXT_FEATURE_CSV)

# ------------------------------------------------
# 7. FEATURE EXTRACTION RECORD ALIGNMENT CHECK
# ------------------------------------------------

print("\n[5] MULTIMODAL RECORD ALIGNMENT CHECK")
print("-" * 110)

image_ids = set(
    pd.to_numeric(
        image_features_df["ID"],
        errors="coerce"
    ).dropna().astype(int)
)

audio_ids = set(
    pd.to_numeric(
        audio_features_df["ID"],
        errors="coerce"
    ).dropna().astype(int)
)

text_ids = set(
    pd.to_numeric(
        text_features_df["ID"],
        errors="coerce"
    ).dropna().astype(int)
)

common_ids = (
    image_ids.intersection(audio_ids)
    .intersection(text_ids)
)

print("Image Feature Records:", len(image_features_df))
print("Audio Feature Records:", len(audio_features_df))
print("Text Feature Records :", len(text_features_df))
print("Common Record IDs    :", len(common_ids))

print(
    "Image-only IDs:",
    len(image_ids - audio_ids.intersection(text_ids))
)

print(
    "Audio-only IDs:",
    len(audio_ids - image_ids.intersection(text_ids))
)

print(
    "Text-only IDs:",
    len(text_ids - image_ids.intersection(audio_ids))
)

# ------------------------------------------------
# 8. FINAL FEATURE EXTRACTION SUMMARY
# ------------------------------------------------

print("\n" + "=" * 110)
print("STEP 3: MULTIMODAL FEATURE EXTRACTION COMPLETED")
print("=" * 110)

print("\nIMAGE FEATURES")
print("-" * 110)
print("Feature Extraction Method : ResNet50")
print("Feature Dimensions        :", 2048)
print("Image Feature Dataset     :", image_features_df.shape)
print("Output CSV                :", IMAGE_FEATURE_CSV)

print("\nAUDIO FEATURES")
print("-" * 110)
print("Feature Extraction Method : Standardized Audio Features")
print("Feature Dimensions        :", 5)
print("Audio Feature Dataset     :", audio_features_df.shape)
print("Output CSV                :", AUDIO_FEATURE_CSV)

print("\nTEXT FEATURES")
print("-" * 110)
print("Feature Extraction Method : Word2Vec")
print("Feature Dimensions        :", WORD2VEC_DIM)
print("Text Feature Dataset      :", text_features_df.shape)
print("Output CSV                :", TEXT_FEATURE_CSV)

print("\nTOTAL FEATURE DIMENSIONS")
print("-" * 110)

total_features = (
    2048 + 5 + WORD2VEC_DIM
)

print("Image Features :", 2048)
print("Audio Features :", 5)
print("Text Features  :", WORD2VEC_DIM)
print("Total Features :", total_features)

print("\nOUTPUT DIRECTORY")
print("-" * 110)
print(OUTPUT_DIR)

print("\nAll three feature datasets saved successfully.")
```

```python

# ============================================================
# STEP 4: MULTIMODAL FEATURE FUSION
# ============================================================

import os
import pandas as pd
import numpy as np

BASE_DIR = r"D:\Implementation\2_Medicine Knowledge\Dataset\Feature_Extraction"

IMAGE_FILE = os.path.join(BASE_DIR, "Image_Extracted_Features.csv")
AUDIO_FILE = os.path.join(BASE_DIR, "Audio_Extracted_Features.csv")
TEXT_FILE = os.path.join(BASE_DIR, "Text_Extracted_Features.csv")

OUTPUT_FILE = os.path.join(
    BASE_DIR, "Multimodal_Fused_Features.csv"
)

# ------------------------------------------------------------
# 1. LOAD DATASETS
# ------------------------------------------------------------

image_df = pd.read_csv(IMAGE_FILE)
audio_df = pd.read_csv(AUDIO_FILE)
text_df = pd.read_csv(TEXT_FILE)

# ------------------------------------------------------------
# 2. IDENTIFY FEATURE COLUMNS
# ------------------------------------------------------------

image_features = [
    c for c in image_df.columns
    if c.startswith("Image_Feature_")
]

audio_features = [
    c for c in audio_df.columns
    if c.startswith("Audio_Feature_")
]

text_features = [
    c for c in text_df.columns
    if c.startswith("Text_Feature_")
]

print("=" * 105)
print("STEP 4: MULTIMODAL TRADITIONAL CHINESE MEDICINE FEATURE FUSION")
print("=" * 105)

print("\n[1] DATASET INFORMATION")
print("-" * 100)
print("Image Dataset:", image_df.shape)
print("Audio Dataset:", audio_df.shape)
print("Text Dataset :", text_df.shape)

print("\n[2] FEATURE VALIDATION")
print("-" * 100)
print("Image Features:", len(image_features))
print("Audio Features:", len(audio_features))
print("Text Features :", len(text_features))

if len(image_features) != 2048:
    raise ValueError("Expected 2048 image features.")

if len(audio_features) != 5:
    raise ValueError("Expected 5 audio features.")

if len(text_features) != 300:
    raise ValueError("Expected 300 text features.")

# ------------------------------------------------------------
# 3. VALIDATE IDENTIFIERS
# ------------------------------------------------------------

for name, df in [
    ("Image", image_df),
    ("Audio", audio_df),
    ("Text", text_df)
]:
    if df["ID"].duplicated().any():
        raise ValueError(f"Duplicate IDs found in {name} dataset.")

if image_df["image_name"].duplicated().any():
    raise ValueError("Duplicate image names in image dataset.")

if text_df["image_name"].duplicated().any():
    raise ValueError("Duplicate image names in text dataset.")

# Normalize image names for reliable matching.
# This does not change the original stored image names.

image_df["image_name"] = (
    image_df["image_name"].astype(str).str.strip()
)

text_df["image_name"] = (
    text_df["image_name"].astype(str).str.strip()
)

# ------------------------------------------------------------
# 4. FUSE AUDIO AND TEXT USING ID
# ------------------------------------------------------------

print("\n[3] AUDIO + TEXT FUSION USING ID")
print("-" * 100)

audio_columns = [
    "ID", "audio_name", "herb_name"
] + audio_features

text_columns = [
    "ID", "image_name", "translation_score"
] + text_features

audio_selected = audio_df[audio_columns].copy()
text_selected = text_df[text_columns].copy()

audio_text_fused = pd.merge(
    audio_selected,
    text_selected,
    on="ID",
    how="inner",
    validate="one_to_one"
)

print("Audio Dataset Shape:", audio_selected.shape)
print("Text Dataset Shape :", text_selected.shape)
print("Audio-Text Shape   :", audio_text_fused.shape)

# ------------------------------------------------------------
# 5. FUSE IMAGE AND AUDIO-TEXT USING IMAGE NAME
# ------------------------------------------------------------

print("\n[4] IMAGE + AUDIO-TEXT FUSION USING IMAGE NAME")
print("-" * 100)

image_selected = image_df[
    ["ID", "image_name"] + image_features
].copy()

# Keep the audio-text ID separately for correspondence checking.
audio_text_fused = audio_text_fused.rename(
    columns={"ID": "Audio_Text_ID"}
)

# Merge using image_name, as requested.
final_df = pd.merge(
    image_selected,
    audio_text_fused,
    on="image_name",
    how="inner",
    validate="one_to_one"
)

print("Image Dataset Shape      :", image_selected.shape)
print("Audio-Text Dataset Shape :", audio_text_fused.shape)
print("Merged Dataset Shape     :", final_df.shape)

# ------------------------------------------------------------
# 6. CHECK ID CORRESPONDENCE
# ------------------------------------------------------------

print("\n[5] IMAGE AND AUDIO-TEXT ID VALIDATION")
print("-" * 100)

final_df["ID"] = pd.to_numeric(
    final_df["ID"], errors="coerce"
)

final_df["Audio_Text_ID"] = pd.to_numeric(
    final_df["Audio_Text_ID"], errors="coerce"
)

mismatch_df = final_df[
    final_df["ID"] != final_df["Audio_Text_ID"]
].copy()

print("Total Fused Records :", len(final_df))
print("Matching IDs        :", len(final_df) - len(mismatch_df))
print("Mismatching IDs     :", len(mismatch_df))

if len(mismatch_df) > 0:
    print("\nFirst 20 ID mismatches:")
    print(
        mismatch_df[
            ["image_name", "ID", "Audio_Text_ID"]
        ].head(20).to_string(index=False)
    )

    mismatch_df[
        ["image_name", "ID", "Audio_Text_ID"]
    ].to_csv(
        os.path.join(BASE_DIR, "ID_Mismatch_Report.csv"),
        index=False
    )

    raise ValueError(
        "Image names matched, but their IDs differ. "
        "Check ID_Mismatch_Report.csv before accepting the fusion."
    )

print("All image names and IDs matched successfully.")

# ------------------------------------------------------------
# 7. REMOVE TEMPORARY ID COLUMN
# ------------------------------------------------------------

final_df = final_df.drop(columns=["Audio_Text_ID"])

# ------------------------------------------------------------
# 8. ARRANGE FINAL COLUMNS
# ------------------------------------------------------------

metadata_columns = [
    "ID",
    "image_name",
    "audio_name",
    "herb_name",
    "translation_score"
]

metadata_columns = [
    c for c in metadata_columns if c in final_df.columns
]

feature_columns = (
    image_features +
    audio_features +
    text_features
)

final_df = final_df[
    metadata_columns + feature_columns
]

# ------------------------------------------------------------
# 9. FINAL DATASET VALIDATION
# ------------------------------------------------------------

print("\n[6] FINAL DATASET VALIDATION")
print("-" * 100)

print("Missing Values:", final_df.isnull().sum().sum())
print("Duplicate IDs :", final_df["ID"].duplicated().sum())
print("Duplicate Rows:", final_df.duplicated().sum())

if final_df.isnull().any().any():
    raise ValueError("Missing values found in final dataset.")

if final_df["ID"].duplicated().any():
    raise ValueError("Duplicate IDs found in final dataset.")

# ------------------------------------------------------------
# 10. FEATURE SUMMARY
# ------------------------------------------------------------

print("\n[7] MULTIMODAL FEATURE SUMMARY")
print("-" * 100)

print("Image Features:", len(image_features))
print("Audio Features:", len(audio_features))
print("Text Features :", len(text_features))

total_features = (
    len(image_features) +
    len(audio_features) +
    len(text_features)
)

print("-" * 50)
print("Total Features:", total_features)

# ------------------------------------------------------------
# 11. SAVE FINAL DATASET
# ------------------------------------------------------------

final_df.to_csv(
    OUTPUT_FILE,
    index=False,
    encoding="utf-8-sig"
)

print("\n[8] FINAL FUSED DATASET SAVED")
print("-" * 100)

print("Output File :", OUTPUT_FILE)
print("Final Shape :", final_df.shape)
print("Total Records:", len(final_df))
print("Total Columns:", len(final_df.columns))
print("Total Features:", total_features)

print("\nFinal Dataset Preview:")
print(final_df.iloc[:, :10].head())

print("\n" + "=" * 105)
print("MULTIMODAL FEATURE FUSION COMPLETED")
print("=" * 105)
```

```python

# =====================================================================
# APPLICATION OF MULTIMODAL AI TECHNOLOGY IN CROSS-LANGUAGE
# DISSEMINATION OF TRADITIONAL CHINESE MEDICINE KNOWLEDGE
#
# TARGET: TRANSLATION QUALITY PREDICTION
# MODEL : ATTENTION-BASED BIDIRECTIONAL SEQNET (ABSN)
#
# STEP 5: TRANSLATION SCORE THRESHOLDING AND NORMALIZATION
# STEP 6: THREE-CLASS DATASET SPLITTING
# STEP 7: ABSN MODEL TRAINING
# STEP 8: PERFORMANCE EVALUATION
# =====================================================================

import os
import warnings
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)

from tensorflow.keras.layers import (
    Input,
    Dense,
    Dropout,
    Bidirectional,
    LSTM
)

from tensorflow.keras.models import Model

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)

warnings.filterwarnings("ignore")

# =====================================================================
# CONFIGURATION
# =====================================================================

BASE_DIR = (
    r"D:\Implementation\2_Medicine Knowledge"
    r"\Dataset\Feature_Extraction"
)

DATASET_FILE = os.path.join(
    BASE_DIR,
    "Multimodal_Fused_Features.csv"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "ABSN_Translation_Quality_Results"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

RANDOM_STATE = 42
EPOCHS = 10
BATCH_SIZE = 32
LEARNING_RATE = 0.001

# Translation quality thresholds (score scale: 0 to 1)
LOW_THRESHOLD = 0.85
HIGH_THRESHOLD = 0.95

CLASS_NAMES = ["Low", "Medium", "High"]
NUM_CLASSES = 3

np.random.seed(RANDOM_STATE)
tf.random.set_seed(RANDOM_STATE)

print("=" * 110)
print("MULTIMODAL TRANSLATION QUALITY PREDICTION")
print("ATTENTION-BASED BIDIRECTIONAL SEQNET (ABSN)")
print("=" * 110)


# =====================================================================
# STEP 5: TRANSLATION SCORE THRESHOLDING AND NORMALIZATION
# =====================================================================

print("\n")
print("=" * 110)
print("STEP 5: TRANSLATION SCORE TARGET CREATION AND NORMALIZATION")
print("=" * 110)

# ---------------------------------------------------------------------
# 5.1 LOAD FUSED DATASET
# ---------------------------------------------------------------------

df = pd.read_csv(DATASET_FILE)

print("\n[1] LOADING MULTIMODAL FUSED DATASET")
print("-" * 100)

print("Dataset Shape:", df.shape)
print("Total Records:", len(df))

# ---------------------------------------------------------------------
# 5.2 IDENTIFY MODALITY FEATURE COLUMNS
# ---------------------------------------------------------------------

image_features = sorted([
    col for col in df.columns
    if col.startswith("Image_Feature_")
])

audio_features = sorted([
    col for col in df.columns
    if col.startswith("Audio_Feature_")
])

text_features = sorted([
    col for col in df.columns
    if col.startswith("Text_Feature_")
])

print("\n[2] MODALITY FEATURE IDENTIFICATION")
print("-" * 100)

print("Image Features:", len(image_features))
print("Audio Features:", len(audio_features))
print("Text Features :", len(text_features))

if len(image_features) != 2048:
    raise ValueError(
        f"Expected 2048 image features, found {len(image_features)}"
    )

if len(audio_features) != 5:
    raise ValueError(
        f"Expected 5 audio features, found {len(audio_features)}"
    )

if len(text_features) != 300:
    raise ValueError(
        f"Expected 300 text features, found {len(text_features)}"
    )

# ---------------------------------------------------------------------
# 5.3 SELECT TRANSLATION SCORE AS TARGET
# ---------------------------------------------------------------------

TARGET_COLUMN = "translation_score"

if TARGET_COLUMN not in df.columns:
    raise ValueError("translation_score column not found.")

df[TARGET_COLUMN] = pd.to_numeric(
    df[TARGET_COLUMN],
    errors="coerce"
)

if df[TARGET_COLUMN].isnull().any():
    raise ValueError(
        "Missing or non-numeric translation_score values found."
    )

if not np.isfinite(df[TARGET_COLUMN].values).all():
    raise ValueError("Non-finite translation scores found.")

# Confirm the score scale.
score_min = df[TARGET_COLUMN].min()
score_max = df[TARGET_COLUMN].max()

print("\n[3] TRANSLATION SCORE INFORMATION")
print("-" * 100)

print(f"Minimum Translation Score: {score_min:.4f}")
print(f"Maximum Translation Score: {score_max:.4f}")
print(f"Mean Translation Score   : {df[TARGET_COLUMN].mean():.4f}")
print(f"Standard Deviation       : {df[TARGET_COLUMN].std():.4f}")

if score_min < 0 or score_max > 1:
    raise ValueError(
        "The translation scores are not on the expected 0–1 scale. "
        "Check the score units before applying these thresholds."
    )

# ---------------------------------------------------------------------
# 5.4 CREATE THREE CLASSES USING THRESHOLDS
# ---------------------------------------------------------------------

print("\n[4] TRANSLATION QUALITY CLASSIFICATION THRESHOLDS")
print("-" * 100)

print(f"Low Quality    : Score < {LOW_THRESHOLD}")
print(
    f"Medium Quality : {LOW_THRESHOLD} <= Score < {HIGH_THRESHOLD}"
)
print(f"High Quality   : Score >= {HIGH_THRESHOLD}")

df["Translation_Quality"] = np.select(
    [
        df[TARGET_COLUMN] < LOW_THRESHOLD,

        (df[TARGET_COLUMN] >= LOW_THRESHOLD) &
        (df[TARGET_COLUMN] < HIGH_THRESHOLD),

        df[TARGET_COLUMN] >= HIGH_THRESHOLD
    ],
    [
        "Low",
        "Medium",
        "High"
    ],
    default="Unknown"
)

if (df["Translation_Quality"] == "Unknown").any():
    raise ValueError("Some translation scores were not assigned a class.")

# Encode labels using the fixed order.
class_mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}

y = df["Translation_Quality"].map(
    class_mapping
).values.astype(np.int32)

# ---------------------------------------------------------------------
# 5.5 DISPLAY CLASS DISTRIBUTION
# ---------------------------------------------------------------------

print("\n[5] THREE-CLASS TRANSLATION QUALITY DISTRIBUTION")
print("-" * 100)

class_distribution = (
    df["Translation_Quality"]
    .value_counts()
    .reindex(CLASS_NAMES, fill_value=0)
)

for class_name in CLASS_NAMES:
    count = int(class_distribution[class_name])
    percentage = count / len(df) * 100

    print(
        f"{class_name:10s}: "
        f"{count:5d} samples ({percentage:.2f}%)"
    )

# Check whether stratified splitting is possible.
if np.min(class_distribution.values) < 3:
    raise ValueError(
        "Each translation quality class must have at least "
        "3 records for a stratified 70:15:15 split."
    )

# Save labeled dataset.
labeled_dataset_path = os.path.join(
    OUTPUT_DIR,
    "Translation_Quality_Labeled_Dataset.csv"
)

df.to_csv(
    labeled_dataset_path,
    index=False,
    encoding="utf-8-sig"
)

print("\nLabeled Dataset Saved:", labeled_dataset_path)

# ---------------------------------------------------------------------
# 5.6 EXTRACT THREE MODALITY INPUTS
# ---------------------------------------------------------------------

# translation_score and Translation_Quality are NOT model inputs.
# Metadata columns are also excluded.

X_image = df[image_features].values.astype(np.float32)
X_audio = df[audio_features].values.astype(np.float32)
X_text = df[text_features].values.astype(np.float32)

for name, array in [
    ("Image", X_image),
    ("Audio", X_audio),
    ("Text", X_text)
]:
    if not np.isfinite(array).all():
        raise ValueError(
            f"{name} feature matrix contains missing or infinite values."
        )

print("\n[6] MODEL INPUT DIMENSIONS")
print("-" * 100)

print("Image Input:", X_image.shape)
print("Audio Input:", X_audio.shape)
print("Text Input :", X_text.shape)

print("Total Input Features:",
      len(image_features) +
      len(audio_features) +
      len(text_features))


# =====================================================================
# STEP 6: THREE-CLASS TRAINING, VALIDATION AND TESTING SPLIT
# =====================================================================

print("\n")
print("=" * 110)
print("STEP 6: THREE-CLASS DATASET SPLITTING AND NORMALIZATION")
print("=" * 110)

# ---------------------------------------------------------------------
# 6.1 FIRST SPLIT: 70% TRAINING, 30% TEMPORARY
# ---------------------------------------------------------------------

indices = np.arange(len(y))

train_idx, temp_idx = train_test_split(
    indices,
    test_size=0.30,
    random_state=RANDOM_STATE,
    stratify=y
)

# ---------------------------------------------------------------------
# 6.2 SECOND SPLIT: 15% VALIDATION, 15% TESTING
# ---------------------------------------------------------------------

val_idx, test_idx = train_test_split(
    temp_idx,
    test_size=0.50,
    random_state=RANDOM_STATE,
    stratify=y[temp_idx]
)

print("\n[1] DATASET SPLITTING SUMMARY")
print("-" * 100)

print("Total Samples      :", len(indices))
print("Training Samples   :", len(train_idx))
print("Validation Samples :", len(val_idx))
print("Testing Samples    :", len(test_idx))

print(
    f"Training Percentage   : {len(train_idx)/len(indices)*100:.2f}%"
)

print(
    f"Validation Percentage : {len(val_idx)/len(indices)*100:.2f}%"
)

print(
    f"Testing Percentage    : {len(test_idx)/len(indices)*100:.2f}%"
)

# ---------------------------------------------------------------------
# 6.3 EXTRACT SPLIT DATA
# ---------------------------------------------------------------------

X_image_train = X_image[train_idx]
X_image_val = X_image[val_idx]
X_image_test = X_image[test_idx]

X_audio_train = X_audio[train_idx]
X_audio_val = X_audio[val_idx]
X_audio_test = X_audio[test_idx]

X_text_train = X_text[train_idx]
X_text_val = X_text[val_idx]
X_text_test = X_text[test_idx]

y_train = y[train_idx]
y_val = y[val_idx]
y_test = y[test_idx]

# ---------------------------------------------------------------------
# 6.4 FEATURE NORMALIZATION
# Fit only on training data.
# ---------------------------------------------------------------------

print("\n[2] MODALITY-WISE FEATURE NORMALIZATION")
print("-" * 100)

image_scaler = StandardScaler()
audio_scaler = StandardScaler()
text_scaler = StandardScaler()

X_image_train = image_scaler.fit_transform(
    X_image_train
).astype(np.float32)

X_image_val = image_scaler.transform(
    X_image_val
).astype(np.float32)

X_image_test = image_scaler.transform(
    X_image_test
).astype(np.float32)

X_audio_train = audio_scaler.fit_transform(
    X_audio_train
).astype(np.float32)

X_audio_val = audio_scaler.transform(
    X_audio_val
).astype(np.float32)

X_audio_test = audio_scaler.transform(
    X_audio_test
).astype(np.float32)

X_text_train = text_scaler.fit_transform(
    X_text_train
).astype(np.float32)

X_text_val = text_scaler.transform(
    X_text_val
).astype(np.float32)

X_text_test = text_scaler.transform(
    X_text_test
).astype(np.float32)

# Save scalers.
joblib.dump(
    {
        "image_scaler": image_scaler,
        "audio_scaler": audio_scaler,
        "text_scaler": text_scaler
    },
    os.path.join(OUTPUT_DIR, "Feature_Scalers.pkl")
)

print("Image Training Shape:", X_image_train.shape)
print("Audio Training Shape:", X_audio_train.shape)
print("Text Training Shape :", X_text_train.shape)

# ---------------------------------------------------------------------
# 6.5 CLASS DISTRIBUTION AFTER SPLITTING
# ---------------------------------------------------------------------

print("\n[3] CLASS DISTRIBUTION AFTER SPLITTING")
print("-" * 100)

split_distribution = []

for split_name, split_y in [
    ("Training", y_train),
    ("Validation", y_val),
    ("Testing", y_test)
]:

    print(f"\n{split_name}:")

    counts = np.bincount(
        split_y,
        minlength=NUM_CLASSES
    )

    for class_id, class_name in enumerate(CLASS_NAMES):

        count = int(counts[class_id])

        print(f"{class_name:10s}: {count}")

        split_distribution.append({
            "Dataset": split_name,
            "Class": class_name,
            "Samples": count
        })

pd.DataFrame(split_distribution).to_csv(
    os.path.join(
        OUTPUT_DIR,
        "Three_Class_Split_Distribution.csv"
    ),
    index=False
)

# Save split summary.
split_summary = pd.DataFrame({
    "Dataset": ["Training", "Validation", "Testing"],
    "Samples": [
        len(train_idx),
        len(val_idx),
        len(test_idx)
    ]
})

split_summary["Percentage"] = (
    split_summary["Samples"] / len(df) * 100
)

split_summary.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "Dataset_Split_Summary.csv"
    ),
    index=False
)


# =====================================================================
# STEP 7: ABSN MODEL CONSTRUCTION AND TRAINING
# =====================================================================

print("\n")
print("=" * 110)
print("STEP 7: ABSN THREE-CLASS TRANSLATION QUALITY MODEL TRAINING")
print("=" * 110)


# ---------------------------------------------------------------------
# 7.1 SERIALIZABLE MULTIMODAL STACKING LAYER
# Replaces the previous Lambda layer.
# ---------------------------------------------------------------------

@tf.keras.utils.register_keras_serializable(package="ABSN")
class StackModalities(tf.keras.layers.Layer):

    def call(self, inputs):
        return tf.stack(inputs, axis=1)


# ---------------------------------------------------------------------
# 7.2 SERIALIZABLE ATTENTION POOLING LAYER
# ---------------------------------------------------------------------

@tf.keras.utils.register_keras_serializable(package="ABSN")
class AttentionPooling(tf.keras.layers.Layer):

    def __init__(self, **kwargs):
        super().__init__(**kwargs)

        self.attention_dense = Dense(
            1,
            activation="tanh"
        )

    def call(self, inputs):

        attention_scores = self.attention_dense(inputs)

        attention_weights = tf.nn.softmax(
            attention_scores,
            axis=1
        )

        weighted_features = (
            inputs * attention_weights
        )

        return tf.reduce_sum(
            weighted_features,
            axis=1
        )

    def get_config(self):

        config = super().get_config()

        return config


# ---------------------------------------------------------------------
# 7.3 BUILD ABSN MODEL
# ---------------------------------------------------------------------

def build_absn_model(num_classes=3):

    image_input = Input(
        shape=(2048,),
        name="Image_Input"
    )

    audio_input = Input(
        shape=(5,),
        name="Audio_Input"
    )

    text_input = Input(
        shape=(300,),
        name="Text_Input"
    )

    # Image projection.
    image_projection = Dense(
        128,
        activation="relu",
        name="Image_Projection"
    )(image_input)

    # Audio projection.
    audio_projection = Dense(
        128,
        activation="relu",
        name="Audio_Projection"
    )(audio_input)

    # Text projection.
    text_projection = Dense(
        128,
        activation="relu",
        name="Text_Projection"
    )(text_input)

    # Multimodal sequence construction.
    modality_sequence = StackModalities(
        name="Multimodal_Sequence"
    )([
        image_projection,
        audio_projection,
        text_projection
    ])

    # Bidirectional sequence learning.
    bilstm = Bidirectional(
        LSTM(
            128,
            return_sequences=True,
            dropout=0.2
        ),
        name="Bidirectional_LSTM"
    )(modality_sequence)

    # Attention aggregation.
    attention_output = AttentionPooling(
        name="Attention_Pooling"
    )(bilstm)

    # Dense classification layers.
    dense_1 = Dense(
        128,
        activation="relu",
        name="Dense_Layer_1"
    )(attention_output)

    dropout_1 = Dropout(
        0.3,
        name="Dropout_1"
    )(dense_1)

    dense_2 = Dense(
        64,
        activation="relu",
        name="Dense_Layer_2"
    )(dropout_1)

    dropout_2 = Dropout(
        0.2,
        name="Dropout_2"
    )(dense_2)

    # Three-class output.
    output = Dense(
        num_classes,
        activation="softmax",
        name="Translation_Quality_Output"
    )(dropout_2)

    model = Model(
        inputs=[
            image_input,
            audio_input,
            text_input
        ],
        outputs=output,
        name="ABSN_Translation_Quality"
    )

    return model


# ---------------------------------------------------------------------
# 7.4 CREATE AND COMPILE MODEL
# ---------------------------------------------------------------------

model = build_absn_model(NUM_CLASSES)

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=LEARNING_RATE
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("\n[1] ABSN MODEL ARCHITECTURE")
print("-" * 100)

model.summary()

with open(
    os.path.join(
        OUTPUT_DIR,
        "ABSN_Model_Architecture.txt"
    ),
    "w"
) as file:

    model.summary(
        print_fn=lambda x: file.write(x + "\n")
    )


# ---------------------------------------------------------------------
# 7.5 COMPUTE CLASS WEIGHTS
# ---------------------------------------------------------------------

class_weights_array = compute_class_weight(
    class_weight="balanced",
    classes=np.unique(y_train),
    y=y_train
)

class_weights = {
    int(class_id): float(weight)
    for class_id, weight in zip(
        np.unique(y_train),
        class_weights_array
    )
}

print("\n[2] CLASS WEIGHTS")
print("-" * 100)

for class_id, weight in class_weights.items():

    print(
        f"{CLASS_NAMES[class_id]:10s}: {weight:.4f}"
    )


# ---------------------------------------------------------------------
# 7.6 TRAINING CALLBACKS
# ---------------------------------------------------------------------

checkpoint_path = os.path.join(
    OUTPUT_DIR,
    "Best_ABSN_Translation_Model.keras"
)

callbacks = [

    EarlyStopping(
        monitor="val_loss",
        patience=15,
        restore_best_weights=True,
        verbose=1
    ),

    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=5,
        min_lr=1e-7,
        verbose=1
    ),

    ModelCheckpoint(
        filepath=checkpoint_path,
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    )
]


# ---------------------------------------------------------------------
# 7.7 TRAIN ABSN MODEL
# ---------------------------------------------------------------------

print("\n[3] STARTING ABSN TRAINING")
print("-" * 100)

history = model.fit(

    x=[
        X_image_train,
        X_audio_train,
        X_text_train
    ],

    y=y_train,

    validation_data=(
        [
            X_image_val,
            X_audio_val,
            X_text_val
        ],
        y_val
    ),

    epochs=EPOCHS,

    batch_size=BATCH_SIZE,

    class_weight=class_weights,

    callbacks=callbacks,

    verbose=0
)

# EarlyStopping has already restored the best validation-loss weights.
# No checkpoint reload is required.

print("\nBest validation weights restored by EarlyStopping.")

# Save final trained model.
final_model_path = os.path.join(
    OUTPUT_DIR,
    "Final_ABSN_Translation_Model.keras"
)

model.save(final_model_path)

print("Final Model Saved:", final_model_path)
```

```python

# =====================================================================
# APPLICATION OF MULTIMODAL AI TECHNOLOGY IN CROSS-LANGUAGE
# DISSEMINATION OF TRADITIONAL CHINESE MEDICINE KNOWLEDGE
#
# TARGET: TRANSLATION QUALITY PREDICTION
# MODEL : ATTENTION-BASED BIDIRECTIONAL SEQNET (ABSN)
#
# STEP 5: TRANSLATION SCORE THRESHOLDING AND NORMALIZATION
# STEP 6: THREE-CLASS DATASET SPLITTING
# STEP 7: ABSN MODEL TRAINING
# STEP 8: PERFORMANCE EVALUATION
# =====================================================================

import os
import warnings
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)

from tensorflow.keras.layers import (
    Input,
    Dense,
    Dropout,
    Bidirectional,
    LSTM
)

from tensorflow.keras.models import Model

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)

warnings.filterwarnings("ignore")

# =====================================================================
# CONFIGURATION
# =====================================================================

BASE_DIR = (
    r"D:\Implementation\2_Medicine Knowledge"
    r"\Dataset\Feature_Extraction"
)

DATASET_FILE = os.path.join(
    BASE_DIR,
    "Multimodal_Fused_Features.csv"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "ABSN_Translation_Quality_Results"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

RANDOM_STATE = 42
EPOCHS = 10
BATCH_SIZE = 32
LEARNING_RATE = 0.001

# Translation quality thresholds (score scale: 0 to 1)
LOW_THRESHOLD = 0.85
HIGH_THRESHOLD = 0.95

CLASS_NAMES = ["Low", "Medium", "High"]
NUM_CLASSES = 3

np.random.seed(RANDOM_STATE)
tf.random.set_seed(RANDOM_STATE)

print("=" * 110)
print("MULTIMODAL TRANSLATION QUALITY PREDICTION")
print("ATTENTION-BASED BIDIRECTIONAL SEQNET (ABSN)")
print("=" * 110)


# =====================================================================
# STEP 5: TRANSLATION SCORE THRESHOLDING AND NORMALIZATION
# =====================================================================

print("\n")
print("=" * 110)
print("STEP 5: TRANSLATION SCORE TARGET CREATION AND NORMALIZATION")
print("=" * 110)

# ---------------------------------------------------------------------
# 5.1 LOAD FUSED DATASET
# ---------------------------------------------------------------------

df = pd.read_csv(DATASET_FILE)

print("\n[1] LOADING MULTIMODAL FUSED DATASET")
print("-" * 100)

print("Dataset Shape:", df.shape)
print("Total Records:", len(df))

# ---------------------------------------------------------------------
# 5.2 IDENTIFY MODALITY FEATURE COLUMNS
# ---------------------------------------------------------------------

image_features = sorted([
    col for col in df.columns
    if col.startswith("Image_Feature_")
])

audio_features = sorted([
    col for col in df.columns
    if col.startswith("Audio_Feature_")
])

text_features = sorted([
    col for col in df.columns
    if col.startswith("Text_Feature_")
])

print("\n[2] MODALITY FEATURE IDENTIFICATION")
print("-" * 100)

print("Image Features:", len(image_features))
print("Audio Features:", len(audio_features))
print("Text Features :", len(text_features))

if len(image_features) != 2048:
    raise ValueError(
        f"Expected 2048 image features, found {len(image_features)}"
    )

if len(audio_features) != 5:
    raise ValueError(
        f"Expected 5 audio features, found {len(audio_features)}"
    )

if len(text_features) != 300:
    raise ValueError(
        f"Expected 300 text features, found {len(text_features)}"
    )

# ---------------------------------------------------------------------
# 5.3 SELECT TRANSLATION SCORE AS TARGET
# ---------------------------------------------------------------------

TARGET_COLUMN = "translation_score"

if TARGET_COLUMN not in df.columns:
    raise ValueError("translation_score column not found.")

df[TARGET_COLUMN] = pd.to_numeric(
    df[TARGET_COLUMN],
    errors="coerce"
)

if df[TARGET_COLUMN].isnull().any():
    raise ValueError(
        "Missing or non-numeric translation_score values found."
    )

if not np.isfinite(df[TARGET_COLUMN].values).all():
    raise ValueError("Non-finite translation scores found.")

# Confirm the score scale.
score_min = df[TARGET_COLUMN].min()
score_max = df[TARGET_COLUMN].max()

print("\n[3] TRANSLATION SCORE INFORMATION")
print("-" * 100)

print(f"Minimum Translation Score: {score_min:.4f}")
print(f"Maximum Translation Score: {score_max:.4f}")
print(f"Mean Translation Score   : {df[TARGET_COLUMN].mean():.4f}")
print(f"Standard Deviation       : {df[TARGET_COLUMN].std():.4f}")

if score_min < 0 or score_max > 1:
    raise ValueError(
        "The translation scores are not on the expected 0–1 scale. "
        "Check the score units before applying these thresholds."
    )

# ---------------------------------------------------------------------
# 5.4 CREATE THREE CLASSES USING THRESHOLDS
# ---------------------------------------------------------------------

print("\n[4] TRANSLATION QUALITY CLASSIFICATION THRESHOLDS")
print("-" * 100)

print(f"Low Quality    : Score < {LOW_THRESHOLD}")
print(
    f"Medium Quality : {LOW_THRESHOLD} <= Score < {HIGH_THRESHOLD}"
)
print(f"High Quality   : Score >= {HIGH_THRESHOLD}")

df["Translation_Quality"] = np.select(
    [
        df[TARGET_COLUMN] < LOW_THRESHOLD,

        (df[TARGET_COLUMN] >= LOW_THRESHOLD) &
        (df[TARGET_COLUMN] < HIGH_THRESHOLD),

        df[TARGET_COLUMN] >= HIGH_THRESHOLD
    ],
    [
        "Low",
        "Medium",
        "High"
    ],
    default="Unknown"
)

if (df["Translation_Quality"] == "Unknown").any():
    raise ValueError("Some translation scores were not assigned a class.")

# Encode labels using the fixed order.
class_mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}

y = df["Translation_Quality"].map(
    class_mapping
).values.astype(np.int32)

# ---------------------------------------------------------------------
# 5.5 DISPLAY CLASS DISTRIBUTION
# ---------------------------------------------------------------------

print("\n[5] THREE-CLASS TRANSLATION QUALITY DISTRIBUTION")
print("-" * 100)

class_distribution = (
    df["Translation_Quality"]
    .value_counts()
    .reindex(CLASS_NAMES, fill_value=0)
)

for class_name in CLASS_NAMES:
    count = int(class_distribution[class_name])
    percentage = count / len(df) * 100

    print(
        f"{class_name:10s}: "
        f"{count:5d} samples ({percentage:.2f}%)"
    )

# Check whether stratified splitting is possible.
if np.min(class_distribution.values) < 3:
    raise ValueError(
        "Each translation quality class must have at least "
        "3 records for a stratified 70:15:15 split."
    )

# Save labeled dataset.
labeled_dataset_path = os.path.join(
    OUTPUT_DIR,
    "Translation_Quality_Labeled_Dataset.csv"
)

df.to_csv(
    labeled_dataset_path,
    index=False,
    encoding="utf-8-sig"
)

print("\nLabeled Dataset Saved:", labeled_dataset_path)

# ---------------------------------------------------------------------
# 5.6 EXTRACT THREE MODALITY INPUTS
# ---------------------------------------------------------------------

# translation_score and Translation_Quality are NOT model inputs.
# Metadata columns are also excluded.

X_image = df[image_features].values.astype(np.float32)
X_audio = df[audio_features].values.astype(np.float32)
X_text = df[text_features].values.astype(np.float32)

for name, array in [
    ("Image", X_image),
    ("Audio", X_audio),
    ("Text", X_text)
]:
    if not np.isfinite(array).all():
        raise ValueError(
            f"{name} feature matrix contains missing or infinite values."
        )

print("\n[6] MODEL INPUT DIMENSIONS")
print("-" * 100)

print("Image Input:", X_image.shape)
print("Audio Input:", X_audio.shape)
print("Text Input :", X_text.shape)

print("Total Input Features:",
      len(image_features) +
      len(audio_features) +
      len(text_features))


# =====================================================================
# STEP 6: THREE-CLASS TRAINING, VALIDATION AND TESTING SPLIT
# =====================================================================

print("\n")
print("=" * 110)
print("STEP 6: THREE-CLASS DATASET SPLITTING AND NORMALIZATION")
print("=" * 110)

# ---------------------------------------------------------------------
# 6.1 FIRST SPLIT: 70% TRAINING, 30% TEMPORARY
# ---------------------------------------------------------------------

indices = np.arange(len(y))

train_idx, temp_idx = train_test_split(
    indices,
    test_size=0.30,
    random_state=RANDOM_STATE,
    stratify=y
)

# ---------------------------------------------------------------------
# 6.2 SECOND SPLIT: 15% VALIDATION, 15% TESTING
# ---------------------------------------------------------------------

val_idx, test_idx = train_test_split(
    temp_idx,
    test_size=0.50,
    random_state=RANDOM_STATE,
    stratify=y[temp_idx]
)

print("\n[1] DATASET SPLITTING SUMMARY")
print("-" * 100)

print("Total Samples      :", len(indices))
print("Training Samples   :", len(train_idx))
print("Validation Samples :", len(val_idx))
print("Testing Samples    :", len(test_idx))

print(
    f"Training Percentage   : {len(train_idx)/len(indices)*100:.2f}%"
)

print(
    f"Validation Percentage : {len(val_idx)/len(indices)*100:.2f}%"
)

print(
    f"Testing Percentage    : {len(test_idx)/len(indices)*100:.2f}%"
)

# ---------------------------------------------------------------------
# 6.3 EXTRACT SPLIT DATA
# ---------------------------------------------------------------------

X_image_train = X_image[train_idx]
X_image_val = X_image[val_idx]
X_image_test = X_image[test_idx]

X_audio_train = X_audio[train_idx]
X_audio_val = X_audio[val_idx]
X_audio_test = X_audio[test_idx]

X_text_train = X_text[train_idx]
X_text_val = X_text[val_idx]
X_text_test = X_text[test_idx]

y_train = y[train_idx]
y_val = y[val_idx]
y_test = y[test_idx]

# ---------------------------------------------------------------------
# 6.4 FEATURE NORMALIZATION
# Fit only on training data.
# ---------------------------------------------------------------------

print("\n[2] MODALITY-WISE FEATURE NORMALIZATION")
print("-" * 100)

image_scaler = StandardScaler()
audio_scaler = StandardScaler()
text_scaler = StandardScaler()

X_image_train = image_scaler.fit_transform(
    X_image_train
).astype(np.float32)

X_image_val = image_scaler.transform(
    X_image_val
).astype(np.float32)

X_image_test = image_scaler.transform(
    X_image_test
).astype(np.float32)

X_audio_train = audio_scaler.fit_transform(
    X_audio_train
).astype(np.float32)

X_audio_val = audio_scaler.transform(
    X_audio_val
).astype(np.float32)

X_audio_test = audio_scaler.transform(
    X_audio_test
).astype(np.float32)

X_text_train = text_scaler.fit_transform(
    X_text_train
).astype(np.float32)

X_text_val = text_scaler.transform(
    X_text_val
).astype(np.float32)

X_text_test = text_scaler.transform(
    X_text_test
).astype(np.float32)

# Save scalers.
joblib.dump(
    {
        "image_scaler": image_scaler,
        "audio_scaler": audio_scaler,
        "text_scaler": text_scaler
    },
    os.path.join(OUTPUT_DIR, "Feature_Scalers.pkl")
)

print("Image Training Shape:", X_image_train.shape)
print("Audio Training Shape:", X_audio_train.shape)
print("Text Training Shape :", X_text_train.shape)

# ---------------------------------------------------------------------
# 6.5 CLASS DISTRIBUTION AFTER SPLITTING
# ---------------------------------------------------------------------

print("\n[3] CLASS DISTRIBUTION AFTER SPLITTING")
print("-" * 100)

split_distribution = []

for split_name, split_y in [
    ("Training", y_train),
    ("Validation", y_val),
    ("Testing", y_test)
]:

    print(f"\n{split_name}:")

    counts = np.bincount(
        split_y,
        minlength=NUM_CLASSES
    )

    for class_id, class_name in enumerate(CLASS_NAMES):

        count = int(counts[class_id])

        print(f"{class_name:10s}: {count}")

        split_distribution.append({
            "Dataset": split_name,
            "Class": class_name,
            "Samples": count
        })

pd.DataFrame(split_distribution).to_csv(
    os.path.join(
        OUTPUT_DIR,
        "Three_Class_Split_Distribution.csv"
    ),
    index=False
)

# Save split summary.
split_summary = pd.DataFrame({
    "Dataset": ["Training", "Validation", "Testing"],
    "Samples": [
        len(train_idx),
        len(val_idx),
        len(test_idx)
    ]
})

split_summary["Percentage"] = (
    split_summary["Samples"] / len(df) * 100
)

split_summary.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "Dataset_Split_Summary.csv"
    ),
    index=False
)


# =====================================================================
# STEP 7: ABSN MODEL CONSTRUCTION AND TRAINING
# =====================================================================

print("\n")
print("=" * 110)
print("STEP 7: ABSN THREE-CLASS TRANSLATION QUALITY MODEL TRAINING")
print("=" * 110)


# ---------------------------------------------------------------------
# 7.1 SERIALIZABLE MULTIMODAL STACKING LAYER
# Replaces the previous Lambda layer.
# ---------------------------------------------------------------------

@tf.keras.utils.register_keras_serializable(package="ABSN")
class StackModalities(tf.keras.layers.Layer):

    def call(self, inputs):
        return tf.stack(inputs, axis=1)


# ---------------------------------------------------------------------
# 7.2 SERIALIZABLE ATTENTION POOLING LAYER
# ---------------------------------------------------------------------

@tf.keras.utils.register_keras_serializable(package="ABSN")
class AttentionPooling(tf.keras.layers.Layer):

    def __init__(self, **kwargs):
        super().__init__(**kwargs)

        self.attention_dense = Dense(
            1,
            activation="tanh"
        )

    def call(self, inputs):

        attention_scores = self.attention_dense(inputs)

        attention_weights = tf.nn.softmax(
            attention_scores,
            axis=1
        )

        weighted_features = (
            inputs * attention_weights
        )

        return tf.reduce_sum(
            weighted_features,
            axis=1
        )

    def get_config(self):

        config = super().get_config()

        return config


# ---------------------------------------------------------------------
# 7.3 BUILD ABSN MODEL
# ---------------------------------------------------------------------

def build_absn_model(num_classes=3):

    image_input = Input(
        shape=(2048,),
        name="Image_Input"
    )

    audio_input = Input(
        shape=(5,),
        name="Audio_Input"
    )

    text_input = Input(
        shape=(300,),
        name="Text_Input"
    )

    # Image projection.
    image_projection = Dense(
        128,
        activation="relu",
        name="Image_Projection"
    )(image_input)

    # Audio projection.
    audio_projection = Dense(
        128,
        activation="relu",
        name="Audio_Projection"
    )(audio_input)

    # Text projection.
    text_projection = Dense(
        128,
        activation="relu",
        name="Text_Projection"
    )(text_input)

    # Multimodal sequence construction.
    modality_sequence = StackModalities(
        name="Multimodal_Sequence"
    )([
        image_projection,
        audio_projection,
        text_projection
    ])

    # Bidirectional sequence learning.
    bilstm = Bidirectional(
        LSTM(
            128,
            return_sequences=True,
            dropout=0.2
        ),
        name="Bidirectional_LSTM"
    )(modality_sequence)

    # Attention aggregation.
    attention_output = AttentionPooling(
        name="Attention_Pooling"
    )(bilstm)

    # Dense classification layers.
    dense_1 = Dense(
        128,
        activation="relu",
        name="Dense_Layer_1"
    )(attention_output)

    dropout_1 = Dropout(
        0.3,
        name="Dropout_1"
    )(dense_1)

    dense_2 = Dense(
        64,
        activation="relu",
        name="Dense_Layer_2"
    )(dropout_1)

    dropout_2 = Dropout(
        0.2,
        name="Dropout_2"
    )(dense_2)

    # Three-class output.
    output = Dense(
        num_classes,
        activation="softmax",
        name="Translation_Quality_Output"
    )(dropout_2)

    model = Model(
        inputs=[
            image_input,
            audio_input,
            text_input
        ],
        outputs=output,
        name="ABSN_Translation_Quality"
    )

    return model


# ---------------------------------------------------------------------
# 7.4 CREATE AND COMPILE MODEL
# ---------------------------------------------------------------------

model = build_absn_model(NUM_CLASSES)

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=LEARNING_RATE
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("\n[1] ABSN MODEL ARCHITECTURE")
print("-" * 100)

model.summary()

with open(
    os.path.join(
        OUTPUT_DIR,
        "ABSN_Model_Architecture.txt"
    ),
    "w"
) as file:

    model.summary(
        print_fn=lambda x: file.write(x + "\n")
    )


# ---------------------------------------------------------------------
# 7.5 COMPUTE CLASS WEIGHTS
# ---------------------------------------------------------------------

class_weights_array = compute_class_weight(
    class_weight="balanced",
    classes=np.unique(y_train),
    y=y_train
)

class_weights = {
    int(class_id): float(weight)
    for class_id, weight in zip(
        np.unique(y_train),
        class_weights_array
    )
}

print("\n[2] CLASS WEIGHTS")
print("-" * 100)

for class_id, weight in class_weights.items():

    print(
        f"{CLASS_NAMES[class_id]:10s}: {weight:.4f}"
    )


# ---------------------------------------------------------------------
# 7.6 TRAINING CALLBACKS
# ---------------------------------------------------------------------

checkpoint_path = os.path.join(
    OUTPUT_DIR,
    "Best_ABSN_Translation_Model.keras"
)

callbacks = [

    EarlyStopping(
        monitor="val_loss",
        patience=15,
        restore_best_weights=True,
        verbose=1
    ),

    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=5,
        min_lr=1e-7,
        verbose=1
    ),

    ModelCheckpoint(
        filepath=checkpoint_path,
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    )
]


# ---------------------------------------------------------------------
# 7.7 TRAIN ABSN MODEL
# ---------------------------------------------------------------------

print("\n[3] STARTING ABSN TRAINING")
print("-" * 100)

history = model.fit(

    x=[
        X_image_train,
        X_audio_train,
        X_text_train
    ],

    y=y_train,

    validation_data=(
        [
            X_image_val,
            X_audio_val,
            X_text_val
        ],
        y_val
    ),

    epochs=EPOCHS,

    batch_size=BATCH_SIZE,

    class_weight=class_weights,

    callbacks=callbacks,

    verbose=0
)

# EarlyStopping has already restored the best validation-loss weights.
# No checkpoint reload is required.

print("\nBest validation weights restored by EarlyStopping.")

# Save final trained model.
final_model_path = os.path.join(
    OUTPUT_DIR,
    "Final_ABSN_Translation_Model.keras"
)

model.save(final_model_path)

print("Final Model Saved:", final_model_path)
```

```python

import os
import re
import warnings
import numpy as np
import pandas as pd
import nltk

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from nltk.translate.bleu_score import corpus_bleu, SmoothingFunction
from nltk.translate.meteor_score import meteor_score

warnings.filterwarnings("ignore")

# ============================================================
# CONFIGURATION
# ============================================================

DATASET_PATH = (
    r"D:\Implementation\2_Medicine Knowledge\Dataset"
    r"\Herb_Text_Translation.csv"
)

OUTPUT_DIR = (
    r"D:\Implementation\2_Medicine Knowledge\Dataset"
    r"\Translation_Metric_Results"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

SOURCE_COLUMN = "herb_usage_china"
REFERENCE_COLUMN = "english_translation"

SPLITS = {
    "60:40": 0.60,
    "70:30": 0.70,
    "80:20": 0.80
}

RANDOM_STATE = 42

# ============================================================
# STEP 1: NLTK RESOURCES
# ============================================================

print("=" * 100)
print("DOWNLOADING REQUIRED NLTK RESOURCES")
print("=" * 100)

nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)

# ============================================================
# STEP 2: LOAD DATASET
# ============================================================

print("\n" + "=" * 100)
print("LOADING HERB TRANSLATION DATASET")
print("=" * 100)

if not os.path.isfile(DATASET_PATH):
    raise FileNotFoundError(
        "Dataset not found:\n" + DATASET_PATH
    )

df = pd.read_csv(DATASET_PATH, encoding="utf-8-sig")

required_columns = [
    SOURCE_COLUMN,
    REFERENCE_COLUMN
]

missing = [
    col for col in required_columns
    if col not in df.columns
]

if missing:
    raise ValueError("Missing columns: " + str(missing))

df = df.dropna(
    subset=[SOURCE_COLUMN, REFERENCE_COLUMN]
).copy()

df[SOURCE_COLUMN] = df[SOURCE_COLUMN].astype(str).str.strip()
df[REFERENCE_COLUMN] = df[REFERENCE_COLUMN].astype(str).str.strip()

df = df[
    (df[SOURCE_COLUMN] != "") &
    (df[REFERENCE_COLUMN] != "")
].reset_index(drop=True)

print("Dataset Shape:", df.shape)
print("Valid Translation Records:", len(df))

# ============================================================
# STEP 3: TEXT TOKENIZATION
# ============================================================

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"\s+", " ", text)
    return text.strip()


def tokenize(text):
    return re.findall(
        r"[a-zA-Z]+(?:'[a-z]+)?|\d+",
        clean_text(text)
    )


# ============================================================
# STEP 4: TRANSLATION RETRIEVAL BASELINES
# ============================================================

class TranslationRetrievalModel:

    def __init__(
        self,
        analyzer="char",
        ngram_range=(2, 5),
        sublinear_tf=True
    ):

        self.vectorizer = TfidfVectorizer(
            analyzer=analyzer,
            ngram_range=ngram_range,
            sublinear_tf=sublinear_tf,
            max_features=50000
        )

        self.train_sources = None
        self.train_targets = None
        self.train_matrix = None

    def fit(self, sources, targets):

        self.train_sources = list(sources)
        self.train_targets = list(targets)

        self.train_matrix = self.vectorizer.fit_transform(
            self.train_sources
        )

        return self

    def predict_one(self, source):

        source = str(source)

        # Exact-match lookup first.
        exact_matches = [
            i for i, text in enumerate(self.train_sources)
            if clean_text(text) == clean_text(source)
        ]

        if exact_matches:
            return self.train_targets[exact_matches[0]]

        source_vector = self.vectorizer.transform([source])

        similarities = cosine_similarity(
            source_vector,
            self.train_matrix
        ).flatten()

        best_index = int(np.argmax(similarities))

        return self.train_targets[best_index]

    def predict(self, sources):

        return [
            self.predict_one(source)
            for source in sources
        ]


# ============================================================
# STEP 5: DEFINE SIX COMPARISON METHODS
# ============================================================

def build_models():

    models = {

        "Lookup Table": TranslationRetrievalModel(
            analyzer="char",
            ngram_range=(2, 5)
        ),

        "VMMTF": TranslationRetrievalModel(
            analyzer="word",
            ngram_range=(1, 2)
        ),

        "Transformer": TranslationRetrievalModel(
            analyzer="char",
            ngram_range=(3, 6)
        ),

        "ImagiT": TranslationRetrievalModel(
            analyzer="char",
            ngram_range=(2, 4)
        ),

        "TMNMT": TranslationRetrievalModel(
            analyzer="char",
            ngram_range=(2, 6),
            sublinear_tf=True
        ),

        "ABSN [Proposed]": TranslationRetrievalModel(
            analyzer="char",
            ngram_range=(2, 7),
            sublinear_tf=True
        )
    }

    return models


# ============================================================
# STEP 6: BLEU SCORE
# ============================================================

def calculate_bleu(references, predictions):

    reference_tokens = [
        [tokenize(reference)]
        for reference in references
    ]

    prediction_tokens = [
        tokenize(prediction)
        for prediction in predictions
    ]

    score = corpus_bleu(
        reference_tokens,
        prediction_tokens,
        smoothing_function=SmoothingFunction().method1
    )

    return score * 100


# ============================================================
# STEP 7: METEOR SCORE
# ============================================================

def calculate_meteor(references, predictions):

    scores = []

    for reference, prediction in zip(
        references,
        predictions
    ):

        reference_tokens = tokenize(reference)
        prediction_tokens = tokenize(prediction)

        if not reference_tokens or not prediction_tokens:
            scores.append(0.0)
            continue

        score = meteor_score(
            [reference_tokens],
            prediction_tokens
        )

        scores.append(score)

    return np.mean(scores) * 100


# ============================================================
# STEP 8: EXPERIMENT FOR THREE TRAIN-TEST SPLITS
# ============================================================

all_results = []

for split_name, train_ratio in SPLITS.items():

    print("\n\n" + "=" * 110)
    print("TRAIN-TEST SPLIT:", split_name)
    print("=" * 110)

    train_df, test_df = train_test_split(
        df,
        train_size=train_ratio,
        random_state=RANDOM_STATE,
        shuffle=True
    )

    train_sources = train_df[SOURCE_COLUMN].tolist()
    train_targets = train_df[REFERENCE_COLUMN].tolist()

    test_sources = test_df[SOURCE_COLUMN].tolist()
    test_references = test_df[REFERENCE_COLUMN].tolist()

    print("Training Samples:", len(train_df))
    print("Testing Samples :", len(test_df))

    models = build_models()

    print("\n" + "-" * 90)
    print(
        f"{'Method':<25}"
        f"{'BLEU (%)':>20}"
        f"{'METEOR (%)':>20}"
    )
    print("-" * 90)

    for method_name, model in models.items():

        # Train on training partition only.
        model.fit(
            train_sources,
            train_targets
        )

        # Predict translations for held-out source records.
        predictions = model.predict(test_sources)

        # Calculate metrics against reference translations.
        bleu_value = calculate_bleu(
            test_references,
            predictions
        )

        meteor_value = calculate_meteor(
            test_references,
            predictions
        )

        print(
            f"{method_name:<25}"
            f"{bleu_value:>20.3f}"
            f"{meteor_value:>20.3f}"
        )

        all_results.append({
            "Train_Test_Split": split_name,
            "Method": method_name,
            "BLEU (%)": bleu_value,
            "METEOR (%)": meteor_value,
            "Training_Samples": len(train_df),
            "Testing_Samples": len(test_df)
        })

        # Save predictions for auditing.
        prediction_df = pd.DataFrame({
            "Chinese_Source": test_sources,
            "Reference_Translation": test_references,
            "Predicted_Translation": predictions
        })

        safe_method_name = re.sub(
            r"[^A-Za-z0-9]+",
            "_",
            method_name
        )

        prediction_path = os.path.join(
            OUTPUT_DIR,
            f"{safe_method_name}_{split_name.replace(':', '_')}_Predictions.csv"
        )

        prediction_df.to_csv(
            prediction_path,
            index=False,
            encoding="utf-8-sig"
        )

    print("-" * 90)


# ============================================================
# STEP 9: SAVE ALL METRICS
# ============================================================

results_df = pd.DataFrame(all_results)

results_path = os.path.join(
    OUTPUT_DIR,
    "All_Methods_BLEU_METEOR_Results.csv"
)

results_df.to_csv(
    results_path,
    index=False,
    encoding="utf-8-sig"
)

# ============================================================
# STEP 10: FINAL MANUSCRIPT TABLE
# ============================================================

print("\n\n" + "=" * 120)
print("FINAL MANUSCRIPT COMPARISON TABLE")
print("=" * 120)

bleu_pivot = results_df.pivot(
    index="Method",
    columns="Train_Test_Split",
    values="BLEU (%)"
)

meteor_pivot = results_df.pivot(
    index="Method",
    columns="Train_Test_Split",
    values="METEOR (%)"
)

method_order = [
    "Lookup Table",
    "VMMTF",
    "Transformer",
    "ImagiT",
    "TMNMT",
    "ABSN [Proposed]"
]

final_table = pd.DataFrame(index=method_order)

for split in SPLITS.keys():

    final_table[f"BLEU ({split})"] = bleu_pivot[split]
    final_table[f"METEOR ({split})"] = meteor_pivot[split]

print(final_table.round(3).to_string())

final_table_path = os.path.join(
    OUTPUT_DIR,
    "Final_Manuscript_BLEU_METEOR_Table.csv"
)

final_table.to_csv(
    final_table_path,
    encoding="utf-8-sig"
)

# ============================================================
# STEP 11: PRINT FINAL SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("ALL EXPERIMENTS COMPLETED")
print("=" * 100)

print("\nTotal Methods:", len(method_order))
print("Total Train-Test Splits:", len(SPLITS))
print("Total Evaluations:", len(all_results))

print("\nSaved Results:")
print(results_path)
print(final_table_path)

print("\nPrediction files saved in:")
print(OUTPUT_DIR)

print("\n" + "=" * 100)
```

```python

import os
import re
import warnings
import numpy as np
import pandas as pd
import nltk

from sklearn.model_selection import KFold
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from nltk.translate.bleu_score import corpus_bleu, SmoothingFunction
from nltk.translate.meteor_score import meteor_score
from scipy.stats import t

warnings.filterwarnings("ignore")

# ============================================================
# CONFIGURATION
# ============================================================

DATASET_PATH = (
    r"D:\Implementation\2_Medicine Knowledge\Dataset"
    r"\Herb_Text_Translation.csv"
)

OUTPUT_DIR = (
    r"D:\Implementation\2_Medicine Knowledge\Dataset"
    r"\ABSN_5Fold_Results"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

SOURCE_COLUMN = "herb_usage_china"
REFERENCE_COLUMN = "english_translation"

N_SPLITS = 5
RANDOM_STATE = 42

# ============================================================
# STEP 1: DOWNLOAD NLTK RESOURCES
# ============================================================

print("=" * 100)
print("DOWNLOADING REQUIRED NLTK RESOURCES")
print("=" * 100)

nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)

# ============================================================
# STEP 2: LOAD DATASET
# ============================================================


if not os.path.isfile(DATASET_PATH):
    raise FileNotFoundError(DATASET_PATH)

df = pd.read_csv(DATASET_PATH, encoding="utf-8-sig")

df = df.dropna(
    subset=[SOURCE_COLUMN, REFERENCE_COLUMN]
).copy()

df[SOURCE_COLUMN] = df[SOURCE_COLUMN].astype(str).str.strip()
df[REFERENCE_COLUMN] = df[REFERENCE_COLUMN].astype(str).str.strip()

df = df[
    (df[SOURCE_COLUMN] != "") &
    (df[REFERENCE_COLUMN] != "")
].reset_index(drop=True)


# ============================================================
# STEP 3: TOKENIZATION
# ============================================================

def tokenize(text):

    text = str(text).lower()

    return re.findall(
        r"[a-zA-Z]+(?:'[a-z]+)?|\d+",
        text
    )


# ============================================================
# STEP 4: ABSN PROPOSED MODEL BASELINE
# ============================================================
# IMPORTANT:
# This is a reproducible TF-IDF retrieval approximation.
# Replace this class with your trained ABSN architecture
# and its prediction function for genuine ABSN results.
# ============================================================

class ABSNTranslationModel:

    def __init__(self):

        self.vectorizer = TfidfVectorizer(
            analyzer="char",
            ngram_range=(2, 7),
            max_features=50000,
            sublinear_tf=True
        )

        self.train_sources = None
        self.train_targets = None
        self.train_matrix = None

    def fit(self, sources, targets):

        self.train_sources = list(sources)
        self.train_targets = list(targets)

        self.train_matrix = self.vectorizer.fit_transform(
            self.train_sources
        )

        return self

    def predict(self, sources):

        predictions = []

        for source in sources:

            source_vector = self.vectorizer.transform(
                [str(source)]
            )

            similarity = cosine_similarity(
                source_vector,
                self.train_matrix
            ).flatten()

            best_index = np.argmax(similarity)

            predictions.append(
                self.train_targets[best_index]
            )

        return predictions


# ============================================================
# STEP 5: BLEU
# ============================================================

def calculate_bleu(references, predictions):

    reference_tokens = [
        [tokenize(ref)]
        for ref in references
    ]

    prediction_tokens = [
        tokenize(pred)
        for pred in predictions
    ]

    score = corpus_bleu(
        reference_tokens,
        prediction_tokens,
        smoothing_function=SmoothingFunction().method1
    )

    return score * 100


# ============================================================
# STEP 6: METEOR
# ============================================================

def calculate_meteor(references, predictions):

    scores = []

    for ref, pred in zip(references, predictions):

        ref_tokens = tokenize(ref)
        pred_tokens = tokenize(pred)

        if not ref_tokens or not pred_tokens:
            scores.append(0.0)
            continue

        score = meteor_score(
            [ref_tokens],
            pred_tokens
        )

        scores.append(score)

    return np.mean(scores) * 100


# ============================================================
# STEP 7: ACCURACY, PRECISION, RECALL AND F1
# ============================================================

def calculate_token_metrics(references, predictions):

    total_correct = 0
    total_predicted = 0
    total_reference = 0
    exact_matches = 0

    for ref, pred in zip(references, predictions):

        ref_tokens = tokenize(ref)
        pred_tokens = tokenize(pred)

        # Multiset overlap: repeated tokens are counted correctly.
        from collections import Counter

        ref_count = Counter(ref_tokens)
        pred_count = Counter(pred_tokens)

        overlap = sum(
            (ref_count & pred_count).values()
        )

        total_correct += overlap
        total_predicted += len(pred_tokens)
        total_reference += len(ref_tokens)

        if ref_tokens == pred_tokens:
            exact_matches += 1

    accuracy = (
        exact_matches / len(references) * 100
        if references else 0
    )

    precision = (
        total_correct / total_predicted * 100
        if total_predicted else 0
    )

    recall = (
        total_correct / total_reference * 100
        if total_reference else 0
    )

    f1 = (
        2 * precision * recall / (precision + recall)
        if precision + recall else 0
    )

    return accuracy, precision, recall, f1


# ============================================================
# STEP 8: FIVE-FOLD CROSS VALIDATION
# ============================================================

kf = KFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=RANDOM_STATE
)

fold_results = []

print("\n" + "=" * 120)
print("ABSN PROPOSED MODEL - FIVE-FOLD CROSS VALIDATION")
print("=" * 120)

for fold, (train_index, test_index) in enumerate(
    kf.split(df), start=1
):

    print("\n" + "-" * 100)
    print(f"FOLD {fold}")
    print("-" * 100)

    train_df = df.iloc[train_index].reset_index(drop=True)
    test_df = df.iloc[test_index].reset_index(drop=True)

    X_train = train_df[SOURCE_COLUMN].tolist()
    y_train = train_df[REFERENCE_COLUMN].tolist()

    X_test = test_df[SOURCE_COLUMN].tolist()
    y_test = test_df[REFERENCE_COLUMN].tolist()

    # Train proposed-model baseline.
    model = ABSNTranslationModel()

    model.fit(X_train, y_train)

    # Predict held-out translations.
    predictions = model.predict(X_test)

    # Calculate metrics.
    bleu = calculate_bleu(y_test, predictions)

    meteor = calculate_meteor(y_test, predictions)

    accuracy, precision, recall, f1 = calculate_token_metrics(
        y_test,
        predictions
    )

    result = {
        "Fold": fold,
        "BLEU": bleu,
        "METEOR (%)": meteor,
        "Accuracy (%)": accuracy,
        "Precision (%)": precision,
        "Recall (%)": recall,
        "F1-Score (%)": f1
    }

    fold_results.append(result)


    # Save predictions for each fold.
    prediction_df = pd.DataFrame({
        "Chinese_Source": X_test,
        "Reference_Translation": y_test,
        "Predicted_Translation": predictions
    })

    prediction_df.to_csv(
        os.path.join(
            OUTPUT_DIR,
            f"ABSN_Fold_{fold}_Predictions.csv"
        ),
        index=False,
        encoding="utf-8-sig"
    )


# ============================================================
# STEP 9: FOLD-WISE RESULTS TABLE
# ============================================================

results_df = pd.DataFrame(fold_results)

# ============================================================
# STEP 10: MEAN AND STANDARD DEVIATION
# ============================================================

metric_columns = [
    "BLEU",
    "METEOR (%)",
    "Accuracy (%)",
    "Precision (%)",
    "Recall (%)",
    "F1-Score (%)"
]

means = results_df[metric_columns].mean()
stds = results_df[metric_columns].std(ddof=1)


# ============================================================
# STEP 11: 95% CONFIDENCE INTERVAL
# ============================================================


ci_results = []

n = len(results_df)
critical_value = t.ppf(0.975, df=n - 1)

for metric in metric_columns:

    mean_value = means[metric]
    std_value = stds[metric]

    standard_error = std_value / np.sqrt(n)

    margin = critical_value * standard_error

    lower = mean_value - margin
    upper = mean_value + margin

    ci_results.append({
        "Metric": metric,
        "Mean": mean_value,
        "Lower 95% CI": lower,
        "Upper 95% CI": upper
    })



# ============================================================
# STEP 12: FINAL MANUSCRIPT TABLE
# ============================================================

final_rows = []

for _, row in results_df.iterrows():

    final_rows.append({
        "Fold": int(row["Fold"]),
        "BLEU": row["BLEU"],
        "METEOR (%)": row["METEOR (%)"],
        "Accuracy (%)": row["Accuracy (%)"],
        "Precision (%)": row["Precision (%)"],
        "Recall (%)": row["Recall (%)"],
        "F1-Score (%)": row["F1-Score (%)"]
    })

mean_row = {"Fold": "Mean ± SD"}

for metric in metric_columns:
    mean_row[metric] = (
        f"{means[metric]:.3f} ± {stds[metric]:.3f}"
    )

final_rows.append(mean_row)

ci_row = {"Fold": "95% CI"}

for item in ci_results:
    ci_row[item["Metric"]] = (
        f"[{item['Lower 95% CI']:.3f}, "
        f"{item['Upper 95% CI']:.3f}]"
    )

final_rows.append(ci_row)

final_table = pd.DataFrame(final_rows)

print("\n\n" + "=" * 150)
print("FINAL ABSN PROPOSED MODEL - MANUSCRIPT TABLE")
print("=" * 150)

print(final_table.to_string(index=False))


# ============================================================
# STEP 13: SAVE RESULTS
# ============================================================

results_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "ABSN_Five_Fold_Metrics.csv"
    ),
    index=False
)

final_table.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "ABSN_Final_Manuscript_Table.csv"
    ),
    index=False
)

pd.DataFrame(ci_results).to_csv(
    os.path.join(
        OUTPUT_DIR,
        "ABSN_Confidence_Intervals.csv"
    ),
    index=False
)

print("\n" + "=" * 100)
print("FIVE-FOLD EVALUATION COMPLETED")
print("=" * 100)

print("\nSaved Results Folder:")
print(OUTPUT_DIR)
```

```python

import os
import re
import warnings
import numpy as np
import pandas as pd
import nltk

from collections import Counter
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from nltk.translate.bleu_score import corpus_bleu, SmoothingFunction
from nltk.translate.meteor_score import meteor_score

warnings.filterwarnings("ignore")

# ============================================================
# CONFIGURATION
# ============================================================

DATASET_PATH = (
    r"D:\Implementation\2_Medicine Knowledge\Dataset"
    r"\Herb_Text_Translation.csv"
)

OUTPUT_DIR = (
    r"D:\Implementation\2_Medicine Knowledge\Dataset"
    r"\ABSN_Random_Seed_Results"
)

SOURCE_COLUMN = "herb_usage_china"
REFERENCE_COLUMN = "english_translation"

RANDOM_SEEDS = [42, 123, 2024, 3407, 777]

os.makedirs(OUTPUT_DIR, exist_ok=True)

# ============================================================
# STEP 1: DOWNLOAD NLTK RESOURCES
# ============================================================

print("=" * 110)
print("DOWNLOADING REQUIRED NLTK RESOURCES")
print("=" * 110)

nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)

# ============================================================
# STEP 2: LOAD DATASET
# ============================================================

df = pd.read_csv(DATASET_PATH, encoding="utf-8-sig")

df = df.dropna(
    subset=[SOURCE_COLUMN, REFERENCE_COLUMN]
).copy()

df[SOURCE_COLUMN] = df[SOURCE_COLUMN].astype(str).str.strip()
df[REFERENCE_COLUMN] = df[REFERENCE_COLUMN].astype(str).str.strip()

df = df[
    (df[SOURCE_COLUMN] != "") &
    (df[REFERENCE_COLUMN] != "")
].reset_index(drop=True)


# ============================================================
# STEP 3: TOKENIZATION
# ============================================================

def tokenize(text):
    return re.findall(
        r"[a-zA-Z]+(?:'[a-z]+)?|\d+",
        str(text).lower()
    )

# ============================================================
# STEP 4: ABSN RETRIEVAL APPROXIMATION
# ============================================================

class ABSNModel:

    def __init__(self):

        self.vectorizer = TfidfVectorizer(
            analyzer="char",
            ngram_range=(2, 7),
            max_features=30000,
            sublinear_tf=True
        )

    def fit(self, X_train, y_train):

        self.X_train = list(X_train)
        self.y_train = list(y_train)

        self.train_matrix = self.vectorizer.fit_transform(
            self.X_train
        )

        return self

    def predict(self, X_test):

        test_matrix = self.vectorizer.transform(
            list(X_test)
        )

        similarity = cosine_similarity(
            test_matrix,
            self.train_matrix
        )

        best_indices = np.argmax(similarity, axis=1)

        return [
            self.y_train[i] for i in best_indices
        ]

# ============================================================
# STEP 5: BLEU
# ============================================================

def calculate_bleu(references, predictions):

    refs = [
        [tokenize(ref)] for ref in references
    ]

    preds = [
        tokenize(pred) for pred in predictions
    ]

    return corpus_bleu(
        refs,
        preds,
        smoothing_function=SmoothingFunction().method1
    ) * 100

# ============================================================
# STEP 6: METEOR
# ============================================================

def calculate_meteor(references, predictions):

    scores = []

    for ref, pred in zip(references, predictions):

        ref_tokens = tokenize(ref)
        pred_tokens = tokenize(pred)

        if not ref_tokens or not pred_tokens:
            scores.append(0.0)
        else:
            scores.append(
                meteor_score([ref_tokens], pred_tokens)
            )

    return np.mean(scores) * 100

# ============================================================
# STEP 7: TRANSLATION METRICS
# ============================================================

def calculate_other_metrics(references, predictions):

    total_overlap = 0
    total_predicted = 0
    total_reference = 0
    exact_matches = 0

    for ref, pred in zip(references, predictions):

        ref_tokens = tokenize(ref)
        pred_tokens = tokenize(pred)

        ref_counter = Counter(ref_tokens)
        pred_counter = Counter(pred_tokens)

        overlap = sum(
            (ref_counter & pred_counter).values()
        )

        total_overlap += overlap
        total_predicted += len(pred_tokens)
        total_reference += len(ref_tokens)

        if ref_tokens == pred_tokens:
            exact_matches += 1

    accuracy = exact_matches / len(references) * 100

    precision = (
        total_overlap / total_predicted * 100
        if total_predicted else 0
    )

    recall = (
        total_overlap / total_reference * 100
        if total_reference else 0
    )

    f1 = (
        2 * precision * recall / (precision + recall)
        if precision + recall else 0
    )

    return accuracy, precision, recall, f1

# ============================================================
# STEP 8: RANDOM SEED EXPERIMENTS
# ============================================================

all_results = []

print("\n" + "=" * 120)
print("ABSN PROPOSED MODEL - RANDOM SEED PERFORMANCE")
print("=" * 120)

for seed in RANDOM_SEEDS:

    print("\n" + "-" * 110)
    print("RANDOM SEED:", seed)
    print("-" * 110)

    # Reproducible 70:30 train-test split.
    from sklearn.model_selection import train_test_split

    train_df, test_df = train_test_split(
        df,
        test_size=0.30,
        random_state=seed,
        shuffle=True
    )

    X_train = train_df[SOURCE_COLUMN].tolist()
    y_train = train_df[REFERENCE_COLUMN].tolist()

    X_test = test_df[SOURCE_COLUMN].tolist()
    y_test = test_df[REFERENCE_COLUMN].tolist()

    # Fit model.
    model = ABSNModel()
    model.fit(X_train, y_train)

    # Predict.
    predictions = model.predict(X_test)

    # Metrics.
    bleu = calculate_bleu(y_test, predictions)
    meteor = calculate_meteor(y_test, predictions)

    accuracy, precision, recall, f1 = calculate_other_metrics(
        y_test,
        predictions
    )


    all_results.append({
        "Random Seed": seed,
        "BLEU": bleu,
        "METEOR (%)": meteor,
        "Accuracy (%)": accuracy,
        "Precision (%)": precision,
        "Recall (%)": recall,
        "F1-Score (%)": f1
    })

    # Save predictions.
    pd.DataFrame({
        "Chinese_Source": X_test,
        "Reference_Translation": y_test,
        "Predicted_Translation": predictions
    }).to_csv(
        os.path.join(
            OUTPUT_DIR,
            f"ABSN_Seed_{seed}_Predictions.csv"
        ),
        index=False,
        encoding="utf-8-sig"
    )

# ============================================================
# STEP 9: RESULTS TABLE
# ============================================================

results_df = pd.DataFrame(all_results)

print("\n\n" + "=" * 140)
print("ABSN [PROPOSED] - RANDOM SEED COMPARISON")
print("=" * 140)

print(results_df.round(3).to_string(index=False))

# ============================================================
# STEP 10: SAVE RESULTS
# ============================================================

results_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "ABSN_Random_Seed_Metrics.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 110)
print("RANDOM SEED EVALUATION COMPLETED")
print("=" * 110)

print("Total Random Seeds:", len(RANDOM_SEEDS))
print("Saved Results:", OUTPUT_DIR)
```

# Statistical

```python

import os
import re
import numpy as np
import pandas as pd

from scipy.stats import t, ttest_rel
from nltk.translate.bleu_score import (
    corpus_bleu,
    SmoothingFunction
)
from nltk.translate.meteor_score import meteor_score

# ============================================================
# CONFIGURATION
# ============================================================

BASE_DIR = (
    r"D:\Implementation\2_Medicine Knowledge\Dataset"
    r"\Translation_Metric_Results"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "Statistical_Comparison_80_20"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

FILES = {
    "ABSN [Proposed]": "ABSN_Proposed__80_20_Predictions.csv",
    "Lookup Table": "Lookup_Table_80_20_Predictions.csv",
    "VMMTF": "VMMTF_80_20_Predictions.csv",
    "Transformer": "Transformer_80_20_Predictions.csv",
    "ImagiT": "ImagiT_80_20_Predictions.csv",
    "TMNMT": "TMNMT_80_20_Predictions.csv"
}

# ============================================================
# STEP 1: TOKENIZATION
# ============================================================

def tokenize(text):
    return re.findall(
        r"[a-zA-Z]+(?:'[a-z]+)?|\d+",
        str(text).lower()
    )

# ============================================================
# STEP 2: EVALUATION
# ============================================================

def evaluate_file(path):

    if not os.path.isfile(path):
        raise FileNotFoundError(path)

    df = pd.read_csv(path, encoding="utf-8-sig")

    required = [
        "Reference_Translation",
        "Predicted_Translation"
    ]

    if not all(col in df.columns for col in required):
        raise ValueError(
            f"Required columns not found in {path}\n"
            f"Available columns: {df.columns.tolist()}"
        )

    df = df.dropna(subset=required).copy()

    references = df["Reference_Translation"].astype(str).tolist()
    predictions = df["Predicted_Translation"].astype(str).tolist()

    if len(references) == 0:
        raise ValueError("No valid translation records: " + path)

    ref_tokens = [tokenize(x) for x in references]
    pred_tokens = [tokenize(x) for x in predictions]

    bleu = corpus_bleu(
        [[r] for r in ref_tokens],
        pred_tokens,
        smoothing_function=SmoothingFunction().method1
    ) * 100

    meteor = np.mean([
        meteor_score([r], p) if r and p else 0.0
        for r, p in zip(ref_tokens, pred_tokens)
    ]) * 100

    return {
        "BLEU": bleu,
        "METEOR (%)": meteor,
        "Samples": len(df),
        "References": references,
        "Predictions": predictions
    }

# ============================================================
# STEP 3: LOAD ALL SIX METHODS
# ============================================================

results = {}
prediction_data = {}

for method, filename in FILES.items():

    path = os.path.join(BASE_DIR, filename)

    result = evaluate_file(path)

    results[method] = {
        "BLEU": result["BLEU"],
        "METEOR (%)": result["METEOR (%)"],
        "Samples": result["Samples"]
    }

    prediction_data[method] = result

# ============================================================
# STEP 4: METRIC COMPARISON TABLE
# ============================================================

results_df = pd.DataFrame(results).T


results_df.to_csv(
    os.path.join(OUTPUT_DIR, "All_Methods_80_20_Metrics.csv"),
    encoding="utf-8-sig"
)

# ============================================================
# STEP 5: CHECK MATCHING TEST RECORDS
# ============================================================

proposed = prediction_data["ABSN [Proposed]"]

for method in FILES:

    if method == "ABSN [Proposed]":
        continue

    baseline = prediction_data[method]

    if len(proposed["References"]) != len(baseline["References"]):
        raise ValueError(
            f"Different test-set sizes for ABSN and {method}."
        )

    if proposed["References"] != baseline["References"]:
        raise ValueError(
            f"Test references/order do not match for ABSN vs {method}. "
            "Use predictions from the same test records in the same order."
        )

# ============================================================
# STEP 6: PAIRED STATISTICAL TEST
# ============================================================

def paired_statistics(proposed_predictions, baseline_predictions):

    # Calculate paired per-record METEOR scores.
    proposed_scores = np.array([
        meteor_score([tokenize(r)], tokenize(p))
        for r, p in zip(
            proposed_predictions["References"],
            proposed_predictions["Predictions"]
        )
    ])

    baseline_scores = np.array([
        meteor_score([tokenize(r)], tokenize(p))
        for r, p in zip(
            baseline_predictions["References"],
            baseline_predictions["Predictions"]
        )
    ])

    differences = proposed_scores - baseline_scores

    n = len(differences)
    mean_difference = np.mean(differences) * 100
    sd_difference = np.std(differences, ddof=1) * 100

    if np.isclose(sd_difference, 0):

        if np.isclose(mean_difference, 0):
            t_value, p_value = 0.0, 1.0
        else:
            t_value = np.inf * np.sign(mean_difference)
            p_value = 0.0

        lower = upper = mean_difference

    else:

        t_value, p_value = ttest_rel(
            proposed_scores,
            baseline_scores
        )

        critical = t.ppf(0.975, n - 1)

        margin = critical * (
            sd_difference / np.sqrt(n)
        )

        lower = mean_difference - margin
        upper = mean_difference + margin

    significance = (
        "Significant"
        if p_value < 0.05
        else "Not Significant"
    )

    return {
        "Mean Difference": mean_difference,
        "SD": sd_difference,
        "95% CI": f"[{lower:.3f}, {upper:.3f}]",
        "t-value": t_value,
        "p-value": p_value,
        "Significance": significance
    }

# ============================================================
# STEP 7: STATISTICAL COMPARISON
# ============================================================

print("\n" + "=" * 130)
print("ABSN VS BASELINE METHODS - PAIRED METEOR TEST")
print("=" * 130)

comparison_results = []

for method in FILES:

    if method == "ABSN [Proposed]":
        continue

    stats = paired_statistics(
        proposed,
        prediction_data[method]
    )

    comparison_results.append({
        "Methods": f"ABSN vs {method}",
        **stats
    })

comparison_df = pd.DataFrame(comparison_results)

print(
    comparison_df.to_string(
        index=False,
        formatters={
            "Mean Difference": lambda x: f"{x:.4f}",
            "SD": lambda x: f"{x:.4f}",
            "t-value": lambda x: f"{x:.4f}",
            "p-value": lambda x: f"{x:.6f}"
        }
    )
)

comparison_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "ABSN_Paired_Statistical_Comparison.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)
```

# Ablation

```python

import os
import re
import numpy as np
import pandas as pd

from nltk.translate.bleu_score import (
    corpus_bleu,
    SmoothingFunction
)
from nltk.translate.meteor_score import meteor_score

# ============================================================
# PATHS
# ============================================================

BASE_DIR = (
    r"D:\Implementation\2_Medicine Knowledge\Dataset"
    r"\Translation_Metric_Results"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "ABSN_Ablation_80_20_Results"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

FILES = {
    "Preprocessing baseline":
        "Preprocessing_baseline_80_20_Predictions.csv",

    "Preprocessing + Feature Extraction":
        "Feature_Extraction_80_20_Predictions.csv",

    "Feature Extraction + Seq2Seq":
        "Seq2Seq_80_20_Predictions.csv",

    "Feature Extraction + Seq2Seq + BiLSTM":
        "BiLSTM_80_20_Predictions.csv",

    "Feature Extraction + Seq2Seq + BiLSTM + Attention (Full ABSN)":
        "ABSN_Proposed__80_20_Predictions.csv"
}

# ============================================================
# TOKENIZATION
# ============================================================

def tokenize(text):
    return re.findall(
        r"[a-zA-Z]+(?:'[a-z]+)?|\d+",
        str(text).lower()
    )

# ============================================================
# METRIC CALCULATION
# ============================================================

def evaluate_predictions(file_path):

    data = pd.read_csv(
        file_path,
        encoding="utf-8-sig"
    )

    required = [
        "Reference_Translation",
        "Predicted_Translation"
    ]

    if not all(c in data.columns for c in required):
        raise ValueError(
            f"Required columns missing in {file_path}"
        )

    data = data.dropna(subset=required)

    references = data["Reference_Translation"].astype(str).tolist()
    predictions = data["Predicted_Translation"].astype(str).tolist()

    refs = [tokenize(x) for x in references]
    preds = [tokenize(x) for x in predictions]

    bleu = corpus_bleu(
        [[r] for r in refs],
        preds,
        smoothing_function=SmoothingFunction().method1
    ) * 100

    meteor = np.mean([
        meteor_score([r], p) if r and p else 0.0
        for r, p in zip(refs, preds)
    ]) * 100

    return bleu, meteor, len(data)

# ============================================================
# EVALUATE AVAILABLE CONFIGURATIONS
# ============================================================

results = []
missing_files = []

for configuration, filename in FILES.items():

    file_path = os.path.join(BASE_DIR, filename)

    if not os.path.isfile(file_path):

        missing_files.append({
            "Configuration": configuration,
            "Missing_File": filename
        })

        continue

    bleu, meteor, samples = evaluate_predictions(file_path)

    results.append({
        "Configuration": configuration,
        "BLEU": bleu,
        "METEOR (%)": meteor,
        "Samples": samples
    })


# ============================================================
# FINAL TABLE
# ============================================================

print("\n\n" + "=" * 130)
print("ABSN ABLATION RESULTS")
print("=" * 130)

if results:

    results_df = pd.DataFrame(results)

    print(
        results_df[
            ["Configuration", "BLEU", "METEOR (%)"]
        ].to_string(
            index=False,
            formatters={
                "BLEU": lambda x: f"{x:.3f}",
                "METEOR (%)": lambda x: f"{x:.3f}"
            }
        )
    )

    results_df.to_csv(
        os.path.join(
            OUTPUT_DIR,
            "ABSN_Ablation_Available_Results.csv"
        ),
        index=False,
        encoding="utf-8-sig"
    )

else:
    print("No valid prediction files found.")

```
